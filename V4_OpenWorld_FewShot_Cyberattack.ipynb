## Task 1. Resolve exact input files and create a provenance record
The five names may be bare files, files with extensions, or folders. Ambiguous matches raise an error instead of silently choosing a different subset. The fifth file is a feature dictionary; model features are checked from the actual traffic files.

```python
DATASETS = {
    "NB15": "NF-UNSW-NB15-v2",
    "TON": "NF-ToN-IoT-v2",
    "BOT": "NF-BoT-IoT-v2",
    "CIC": "NF-CSE-CIC-IDS2018-v2",
}
EXTS = ('.csv', '.csv.gz', '.parquet', '.zip')

def resolve_input(stem):
    exact = []
    for p in ROOT.rglob('*'):
        if 'research_outputs' in p.parts or p.name.startswith('.'):
            continue
        n = p.name.lower()
        if p.is_file() and any(n == stem.lower() + ext for ext in EXTS): exact.append(p)
        if p.is_dir() and n == stem.lower():
            exact.extend([q for q in p.rglob('*') if q.is_file() and any(q.name.lower().endswith(x) for x in EXTS)])
    exact = sorted(set(exact))
    if len(exact) != 1:
        raise ValueError(f"Expected exactly one file for {stem}, found {len(exact)}: {[str(x) for x in exact]}")
    return exact[0]

INPUTS = {key: resolve_input(stem) for key, stem in DATASETS.items()}
feature_candidates = [p for p in ROOT.rglob('*') if p.is_file() and p.name.lower().startswith('netflow_v2_features')]
FEATURE_DICTIONARY = str(feature_candidates[0]) if len(feature_candidates) == 1 else None

def source_fingerprint(path):
    s = path.stat()
    digest=hashlib.sha256()
    with open(path,'rb') as fh:
        while block:=fh.read(8*1024*1024): digest.update(block)
    return {"name":path.name, "bytes":s.st_size, "sha256":digest.hexdigest(),
            "mtime_utc":datetime.fromtimestamp(s.st_mtime, timezone.utc).isoformat()}

provenance = {"mode":CFG.mode, "created_utc":datetime.now(timezone.utc).isoformat(),
              "settings":asdict(CFG), "python":sys.version, "platform":platform.platform(),
              "packages":{"numpy":np.__version__,"pandas":pd.__version__,"scikit_learn":sklearn.__version__,
                          "torch":torch.__version__},
              "feature_dictionary":FEATURE_DICTIONARY,
              "inputs":{d:source_fingerprint(p) for d,p in INPUTS.items()}}
(OUT/'provenance.json').write_text(json.dumps(provenance, indent=2, default=str))
display(pd.DataFrame([{**{'dataset':k, 'path':str(v)}, **provenance['inputs'][k]} for k,v in INPUTS.items()]))
```

## Task 2. Stream all source files, audit full class counts, and retain a seeded sample
Each row receives an independent random priority within its class. We keep the lowest priorities across **all** chunks, so early chunks do not receive preferential selection. The full row count and label counts are audited. This can take time for the large BoT-IoT file; the capped samples are cached as Parquet. The archive reader assumes one matching CSV per ZIP. No replacement of the publisher's original data is claimed.

```python
LABEL_ALIASES = ('Attack','attack','attack_cat','attack_category','category','Class','class','type')
BINARY_ALIASES = ('Label','label','binary_label','is_attack')
CANONICAL = {'normal':'Benign','benign':'Benign','0':'Benign',
             'dos':'DoS','ddos':'DDoS','backdoor':'Backdoor',
             'reconnaissance':'Reconnaissance','scanning':'Scanning',
             'theft':'Theft','bruteforce':'BruteForce',
             'brute force':'BruteForce','web attacks':'Web Attacks',
             'injection':'Injection','mitm':'MITM','password':'Password',
             'ransomware':'Ransomware','xss':'XSS'}
BAD_FEATURE_TOKENS = ('IPV4_SRC_ADDR','IPV4_DST_ADDR','IP_SRC','IP_DST','SRC_IP','DST_IP',
                      'FLOW_ID','PKSEQID','TIMESTAMP','SRC_ADDR','DST_ADDR','DATASET')

def pick_label(df):
    found = [c for c in LABEL_ALIASES if c in df.columns]
    if not found:
        raise ValueError(f"No multiclass attack label among columns: {list(df.columns)}. Binary Label alone is insufficient.")
    return found[0]

def canonical_label(x,dataset):
    if pd.isna(x): return 'MISSING'
    val = str(x).strip()
    if dataset=='CIC':
        low=val.lower()
        if low.startswith('dos attacks-'): return 'DoS'
        if low.startswith(('ddos attack-','ddos attacks-')): return 'DDoS'
        if low in {'ftp-bruteforce','ssh-bruteforce'}: return 'BruteForce'
        if low in {'brute force -web','brute force -xss','sql injection'}: return 'Web Attacks'
        if low=='infilteration': return 'Infiltration'  # Correct the source spelling only.
        # The parent group sums reconcile to the publisher's NF-CSE-CIC-IDS2018-v2 table.
    return CANONICAL.get(val.lower(), val)

def chunks_from_path(path):
    if path.name.lower().endswith('.parquet'):
        import pyarrow.parquet as pq
        pf = pq.ParquetFile(path)
        for batch in pf.iter_batches(batch_size=CFG.chunk_size): yield batch.to_pandas()
    elif path.name.lower().endswith('.zip'):
        with zipfile.ZipFile(path) as z:
            members = [m for m in z.namelist() if m.lower().endswith('.csv') and not m.startswith('__MACOSX')]
            if len(members) != 1: raise ValueError(f"ZIP {path} has {len(members)} CSV members: {members[:10]}")
            with z.open(members[0]) as f:
                yield from pd.read_csv(f, chunksize=CFG.chunk_size, low_memory=False)
    else:
        yield from pd.read_csv(path, chunksize=CFG.chunk_size, low_memory=False)

def stream_sample(path, key, cap):
    rng = np.random.default_rng(20260928)
    sample_by_class, counts, raw_counts = {}, Counter(), Counter()
    colnames = None
    for chunk_no, chunk in enumerate(chunks_from_path(path), start=1):
        chunk.columns = chunk.columns.astype(str).str.strip()
        if colnames is None: colnames = list(chunk.columns)
        elif colnames != list(chunk.columns): raise ValueError(f"Schema changed between chunks of {path.name}")
        labcol = pick_label(chunk)
        chunk['__raw_label']=chunk[labcol].astype(str).str.strip()
        raw_counts.update(chunk['__raw_label'].value_counts().to_dict())
        chunk['__label'] = chunk[labcol].map(lambda v:canonical_label(v,key))
        counts.update(chunk['__label'].value_counts().to_dict())
        chunk['__priority'] = rng.random(len(chunk))
        chunk['__row_id'] = np.arange((chunk_no-1)*CFG.chunk_size, (chunk_no-1)*CFG.chunk_size+len(chunk), dtype=np.int64)
        for name, part in chunk.groupby('__label', sort=False):
            limited = part.nsmallest(cap, '__priority')
            if name in sample_by_class:
                sample_by_class[name] = pd.concat([sample_by_class[name], limited], ignore_index=True).nsmallest(cap,'__priority')
            else: sample_by_class[name] = limited.copy()
        if chunk_no % 40 == 0: print(key, "chunks:",chunk_no,"rows:",sum(counts.values()))
    if not counts: raise ValueError(f"Empty input: {path}")
    sample = pd.concat(sample_by_class.values(), ignore_index=True).sort_values('__row_id').reset_index(drop=True)
    sample['__dataset'] = key
    return sample, dict(counts), colnames, dict(raw_counts)

SAMPLES, AUDIT = {}, {}
for key, path in INPUTS.items():
    cache = OUT / f'{key}_sample_{CAP}.parquet'
    record = OUT / f'{key}_audit_{CAP}.json'
    # An audit cache is valid only for exactly the current input identity and sample cap.
    if cache.exists() and record.exists():
        old = json.loads(record.read_text())
        if (old.get('fingerprint') == provenance['inputs'][key] and old.get('cap') == CAP
            and old.get('label_map_version')==CFG.label_map_version
            and old.get('audit_cache_version')==CFG.audit_cache_version
            and old.get('raw_label_counts') is not None):
            SAMPLES[key] = pd.read_parquet(cache)
            AUDIT[key] = old
            print('Reused validated cache:',key)
            continue
        # Full raw-label counts cannot be reconstructed from a capped sample.
        # An older audit without independently scanned raw counts must be rescanned.
    # Reuse an earlier full audit only if the input fingerprint, cap and label
    # mapping match. V8 writes its own tree without altering earlier runs.
    reused=False
    for version in ('v7','v6'):
        previous = ROOT / 'research_outputs' / version / CFG.mode.lower()
        previous_cache = previous / cache.name
        previous_record = previous / record.name
        if cache.exists() or not (previous_cache.exists() and previous_record.exists()):
            continue
        old = json.loads(previous_record.read_text())
        if (old.get('fingerprint') == provenance['inputs'][key] and old.get('cap') == CAP
            and old.get('label_map_version') == CFG.label_map_version
            and old.get('audit_cache_version') == CFG.audit_cache_version
            and old.get('raw_label_counts') is not None):
            shutil.copy2(previous_cache, cache)
            shutil.copy2(previous_record, record)
            SAMPLES[key] = pd.read_parquet(cache)
            AUDIT[key] = old
            print(f'Reused fingerprint-verified {version} audit/sample:', key)
            reused=True
            break
    if reused: continue
    frame, counts, cols, raw_counts = stream_sample(path, key, CAP)
    record_data = {"fingerprint":provenance['inputs'][key],"cap":CAP,
                   "total_rows":sum(counts.values()),"full_label_counts":counts,
                   "raw_label_counts":raw_counts,"label_map_version":CFG.label_map_version,
                   "audit_cache_version":CFG.audit_cache_version,
                   "sample_label_counts":frame['__label'].value_counts().to_dict(),"columns":cols}
    frame.to_parquet(cache, index=False, compression='zstd')
    record.write_text(json.dumps(record_data, indent=2))
    SAMPLES[key], AUDIT[key] = frame, record_data
display(pd.DataFrame([{'dataset':k,'total_rows':v['total_rows'],'sample_rows':len(SAMPLES[k]),
                       'classes':len(v['full_label_counts']),'raw_columns':len(v['columns'])}
                      for k,v in AUDIT.items()]))

# Publisher-v2 family totals are an independent check on the CIC subtype mapping.
EXPECTED_CIC={'Benign':16635567,'BruteForce':120912,'Bot':143097,
              'DoS':483999,'DDoS':1390270,'Infiltration':116361,'Web Attacks':3502}
cic_check=pd.DataFrame([{'family':k,'publisher_n':v,
     'observed_n':AUDIT['CIC']['full_label_counts'].get(k,0)} for k,v in EXPECTED_CIC.items()])
cic_check['matches']=cic_check.publisher_n.eq(cic_check.observed_n)
cic_check.to_csv(OUT/'cic_publisher_family_reconciliation.csv',index=False)
if AUDIT['CIC']['total_rows']==18_893_708 and not cic_check.matches.all():
    raise ValueError('CIC family counts differ from the publisher v2 totals; inspect raw labels before proceeding')

if FEATURE_DICTIONARY is None: raise FileNotFoundError('NetFlow_v2_Features dictionary is required for the schema audit')
dictionary=pd.read_csv(FEATURE_DICTIONARY)
if 'Feature' not in dictionary: raise ValueError('Feature dictionary requires a Feature column')
expected_features=set(dictionary['Feature'].astype(str).str.strip())
schema=[]
for key,a in AUDIT.items():
    columns=set(a['columns']);observed_features=columns-{'Attack','Label'}
    times=[c for c in columns if ('START' in c.upper() or 'END' in c.upper()) and
           ('TIME' in c.upper() or 'MILLISECOND' in c.upper())]
    schema.append({'dataset':key,'expected_features':len(expected_features),
       'observed_features':len(observed_features),'missing':json.dumps(sorted(expected_features-observed_features)),
       'extra':json.dumps(sorted(observed_features-expected_features)),
       'verified_time_columns':json.dumps(times)})
    if observed_features!=expected_features: raise ValueError(f'{key} feature columns differ from NetFlow_v2_Features.csv')
pd.DataFrame(schema).to_csv(OUT/'feature_schema_audit.csv',index=False)
display(pd.DataFrame(schema))
```

## Task 3. Audit labels, features, imbalance, duplicates, and group feasibility
We group repeated 41-feature *model signatures* in the capped sample and quarantine signatures with conflicting labels. Endpoint IPs are excluded from the signature: distinct flows can therefore collide, and a collision is not proof of a duplicated or mislabeled source flow. The quarantine sensitivity is essential. This cannot establish that the original entire CSV is duplicate-free. No verified timestamps exist for a prospective chronological test.

```python
def feature_candidates(df):
    excluded = set(LABEL_ALIASES)|set(BINARY_ALIASES)
    return [c for c in df.columns if c not in excluded and not c.startswith('__')
            and not any(t in c.upper() for t in BAD_FEATURE_TOKENS)]

feature_sets = [set(feature_candidates(f)) for f in SAMPLES.values()]
COMMON_FEATURES = sorted(set.intersection(*feature_sets))
if len(COMMON_FEATURES) < 10:
    raise ValueError(f"Only {len(COMMON_FEATURES)} shared non-ID features: {COMMON_FEATURES}. Verify all four files are NetFlow v2.")
print('Shared non-ID feature columns:', len(COMMON_FEATURES))
print(COMMON_FEATURES)

RAW_SAMPLES={k:v.copy() for k,v in SAMPLES.items()}

def clean_and_group(df, dataset, policy='quarantine'):
    x = df.copy()
    x = x[x['__label'].ne('MISSING')].copy()
    # Raw values, excluding labels and identifiers; labels are NEVER part of the signature.
    payload = x[COMMON_FEATURES].astype('string').fillna('<NA>')
    x['__group'] = pd.util.hash_pandas_object(payload, index=False).to_numpy(dtype=np.uint64)
    conflicts = x.groupby('__group')['__label'].nunique()
    bad = set(conflicts[conflicts > 1].index)
    conflicting_rows = int(x['__group'].isin(bad).sum())
    conflicted_by_class=x[x['__group'].isin(bad)]['__label'].value_counts().to_dict()
    if policy=='quarantine':
        x = x[~x['__group'].isin(bad)].drop_duplicates('__group').reset_index(drop=True)
    elif policy=='retain_grouped':
        x=x.reset_index(drop=True)  # All repeats and conflicting labels remain in one indivisible group.
    else: raise ValueError(policy)
    return x, {"dataset":dataset,"policy":policy,"sample_before":len(df),
               "conflicting_rows_present":conflicting_rows,
               "conflicting_rows_removed":conflicting_rows if policy=='quarantine' else 0,
               "conflict_label_counts":conflicted_by_class,
               "unique_after":len(x),"duplicate_rows_removed":max(0,len(df)-conflicting_rows-len(x)),
               "unique_groups":x['__group'].nunique()}

dedup_reports = []
for key in SAMPLES:
    SAMPLES[key], rec = clean_and_group(SAMPLES[key],key)
    dedup_reports.append(rec)
pd.DataFrame(dedup_reports).to_csv(OUT/'sample_dedup_audit.csv',index=False)
pd.DataFrame(dedup_reports).pipe(display)
RETAINED, retention_audit={},[]
for key,frame in RAW_SAMPLES.items():
    RETAINED[key],rec=clean_and_group(frame,key,policy='retain_grouped')
    retention_audit.append(rec)
pd.DataFrame(retention_audit).to_csv(OUT/'conflict_retained_audit.csv',index=False)
loss=[]
for key,frame in RAW_SAMPLES.items():
    before=frame['__label'].value_counts();after=SAMPLES[key]['__label'].value_counts()
    for label in sorted(set(before.index)|set(after.index)):
        loss.append({'dataset':key,'label':label,'sample_before':int(before.get(label,0)),
                     'quarantine_after':int(after.get(label,0)),
                     'removed':int(before.get(label,0)-after.get(label,0))})
pd.DataFrame(loss).to_csv(OUT/'dedup_loss_by_class.csv',index=False)
pd.DataFrame([{'dataset':k,'raw':raw,'harmonized':canonical_label(raw,k),'n':n,
               'taxonomy_reference':'https://staff.itee.uq.edu.au/marius/NIDS_datasets/'}
              for k,a in AUDIT.items() for raw,n in a['raw_label_counts'].items()]).to_csv(
              OUT/'label_harmonization.csv',index=False)

class_table = pd.DataFrame({k:pd.Series(v['full_label_counts']) for k,v in AUDIT.items()}).fillna(0).astype(int)
class_table.to_csv(OUT/'full_class_counts.csv')
display(class_table)
if 'CIC' not in DATASETS or not CFG.run_4th_dataset: print('CIC is present for audit; use CFG.run_4th_dataset to choose experiment scope.')
```

## Task 5. Source-only preprocessing and duplicate-safe partitions
Source-only medians/scales and a source development partition are fitted before any target query is evaluated. Each target class receives a disjoint support pool and query pool. Splitting uses exact feature signatures as groups; it prevents exact repeats from crossing pools, but it is *not* independent capture-session or host validation. For a manuscript, discuss this limitation and add timestamp/session splits if reliable metadata becomes available.

```python
def classwise_group_split(frame, query_fraction, seed, min_left=10, min_right=15):
    lefts, rights=[] , []
    for name, piece in frame.groupby('__label'):
        if len(piece) < min_left+min_right or piece['__group'].nunique() < 2:
            raise ValueError(f"{name}: too few independent sampled records/groups for required split")
        ok = False
        for trial in range(20):
            gs = GroupShuffleSplit(n_splits=1,test_size=query_fraction,random_state=seed+trial)
            l,r = next(gs.split(piece,groups=piece['__group']))
            if len(l)>=min_left and len(r)>=min_right:
                lefts.append(piece.iloc[l]); rights.append(piece.iloc[r]); ok=True; break
        if not ok: raise ValueError(f"{name}: cannot produce disjoint support and query pools")
    left,right = pd.concat(lefts).copy(),pd.concat(rights).copy()
    assert set(left['__group']).isdisjoint(set(right['__group']))
    return left,right

def global_group_split(frame,group_col,query_fraction,seed,min_left,min_right):
    # Required whenever one group has multiple labels (retain policy or source-host grouping).
    classes=set(frame['__label'])
    components=int(frame[group_col].nunique())
    if components<2:
        raise ValueError(f'Only {components} independent {group_col} component; strict split impossible')
    group_counts=frame.groupby('__label')[group_col].nunique()
    missing=group_counts[group_counts<2].to_dict()
    if missing:
        raise ValueError(f'Classes confined to one {group_col} component: {missing}')
    for trial in range(300):
        gs=GroupShuffleSplit(n_splits=1,test_size=query_fraction,random_state=seed+trial)
        l,r=next(gs.split(frame,groups=frame[group_col]))
        left,right=frame.iloc[l],frame.iloc[r]
        lc=left['__label'].value_counts();rc=right['__label'].value_counts()
        if all(lc.get(c,0)>=min_left and rc.get(c,0)>=min_right for c in classes):
            assert set(left[group_col]).isdisjoint(set(right[group_col]))
            return left.copy(),right.copy()
    raise ValueError(f'No viable {group_col} split after 300 fixed trials with '
                     f'per-class minimum support={min_left}, query={min_right}; '
                     f'{components} independent components; report infeasible')

def endpoint_components(frame):
    # Connected components of the undirected communication graph. Every IP address
    # appears in exactly one component; splitting components disjoins both endpoints.
    src=frame['IPV4_SRC_ADDR'].astype('string').fillna('<missing>').to_numpy()
    dst=frame['IPV4_DST_ADDR'].astype('string').fillna('<missing>').to_numpy()
    if (src=='<missing>').any() or (dst=='<missing>').any():
        raise ValueError('Missing endpoint IP prevents fully host-disjoint splitting')
    parent={}
    def root(a):
        parent.setdefault(a,a)
        while parent[a]!=a:
            parent[a]=parent[parent[a]];a=parent[a]
        return a
    for a,b in zip(src,dst): parent[root(a)]=root(b)
    return np.asarray([root(a) for a in src],dtype=object)

class NumericSourcePreprocessor:
    def fit(self, src):
        self.keep,self.dropped=[],{}
        for c in COMMON_FEATURES:
            v = pd.to_numeric(src[c],errors='coerce').replace([np.inf,-np.inf],np.nan)
            fraction = v.notna().mean()
            if fraction >= .98 and v.nunique(dropna=True)>1: self.keep.append(c)
            else: self.dropped[c] = f'numeric_fraction={fraction:.3f}, distinct={v.nunique(dropna=True)}'
        if len(self.keep)<8: raise ValueError(f"Only {len(self.keep)} usable numeric features; inspect encoding/schema.")
        v = self._raw(src)
        self.median = np.nanmedian(v,axis=0)
        self.median = np.where(np.isfinite(self.median),self.median,0)
        v = np.where(np.isfinite(v),v,self.median)
        self.q25,self.q75=np.quantile(v,[.25,.75],axis=0)
        self.scale=np.maximum(self.q75-self.q25,1e-6)
        return self
    def _raw(self,frame):
        return np.column_stack([pd.to_numeric(frame[c],errors='coerce').to_numpy(dtype=np.float64) for c in self.keep])
    def transform(self,frame):
        v=self._raw(frame)
        v=np.where(np.isfinite(v),v,self.median)
        return np.clip((v-self.median)/self.scale,-12,12).astype(np.float32)
    def report(self):
        return {'used_features':self.keep,'dropped_features':self.dropped,
                'median':self.median.tolist(),'iqr':self.scale.tolist()}

def prepare_pair(src,tgt,seed,samples=None,pairs=None,split_mode='feature',write_manifest=True):
    if samples is None: samples=SAMPLES
    if pairs is None: pairs=PAIRS
    row=pairs[(pairs.source==src)&(pairs.target==tgt)].iloc[0]
    if not row.eligible: raise ValueError(f'Pair {src}->{tgt} is ineligible')
    known=list(row.known)
    s=samples[src][samples[src]['__label'].isin(known)].copy()
    t=samples[tgt][samples[tgt]['__label'].isin(known+list(row.unknown_choices))].copy()
    # Remove exact shared feature vectors across datasets before any target split.
    overlap=int(t['__group'].isin(set(s['__group'])).sum())
    t=t[~t['__group'].isin(set(s['__group']))].copy()
    if split_mode in {'host','source_host'}:
        for fr in (s,t):
            if 'IPV4_SRC_ADDR' not in fr or 'IPV4_DST_ADDR' not in fr:
                raise ValueError('Endpoint-IP columns absent in v2 file')
            fr['__split_host']=(endpoint_components(fr) if split_mode=='host' else
                                fr['IPV4_SRC_ADDR'].astype('string').fillna('<missing>'))
            if (fr['__split_host']=='<missing>').any(): raise ValueError('Missing source IP prevents a strict host split')
        split=lambda fr,q,sd,ml,mr:global_group_split(fr,'__split_host',q,sd,ml,mr)
    elif split_mode=='feature' and (samples is RETAINED):
        split=lambda fr,q,sd,ml,mr:global_group_split(fr,'__group',q,sd,ml,mr)
    elif split_mode=='feature':
        split=lambda fr,q,sd,ml,mr:classwise_group_split(fr,q,sd,ml,mr)
    else: raise ValueError(split_mode)
    s_train,s_dev=split(s,CFG.source_dev_fraction,seed,25,15)
    t_support,t_query=split(t,CFG.target_query_fraction,seed+10_000,
                            max(CFG.shots)+2,CFG.min_query_rows)
    for name in known:
        if t_support['__label'].eq(name).sum()<CFG.target_calibration_per_known_class:
            raise ValueError(f'{src}->{tgt}: known class {name} lacks target calibration budget')
    # Freeze a source training cap by class, after split, equal for all compared models.
    kept=[]
    for _,part in s_train.groupby('__label',sort=True):
        kept.append(part.sample(min(len(part),CFG.source_model_cap),random_state=seed))
    s_train=pd.concat(kept,ignore_index=True)
    pre=NumericSourcePreprocessor().fit(s_train)
    if not set(s_train['__group']).isdisjoint(set(s_dev['__group'])): raise AssertionError('Source split leak')
    if not set(t_support['__group']).isdisjoint(set(t_query['__group'])): raise AssertionError('Target split leak')
    manifest={'source':src,'target':tgt,'seed':seed,'known':known,'split_mode':split_mode,
              'conflict_policy':'retain_grouped' if samples is RETAINED else 'quarantine',
              'unknown':list(row.unknown_choices), 'train_source_rows':len(s_train),
              'source_dev_rows':len(s_dev),'target_support_rows':len(t_support),
              'target_query_rows':len(t_query),'preprocessor':pre.report(),
              'cross_dataset_duplicate_target_rows_removed':overlap,
              'split_group':('both-endpoint connected component' if split_mode=='host' else
                             'source-IP' if split_mode=='source_host' else 'exact feature signature')}
    if split_mode in {'host','source_host'}:
        assert set(s_train.__split_host).isdisjoint(set(s_dev.__split_host))
        assert set(t_support.__split_host).isdisjoint(set(t_query.__split_host))
        if 'IPV4_DST_ADDR' in t:
            both=lambda fr:set(fr['IPV4_SRC_ADDR'].astype(str))|set(fr['IPV4_DST_ADDR'].astype(str))
            manifest['target_any_endpoint_overlap']=len(both(t_support)&both(t_query))
            manifest['full_endpoint_host_disjoint']=manifest['target_any_endpoint_overlap']==0
            if split_mode=='host': assert manifest['full_endpoint_host_disjoint']
    if write_manifest:
        stem=f'{src}_to_{tgt}_seed{seed}_{split_mode}'
        (OUT/f'{stem}_protocol.json').write_text(json.dumps(manifest,indent=2))
        for name,fr in [('source_train',s_train),('source_dev',s_dev),('target_support',t_support),('target_query',t_query)]:
            fr[['__dataset','__row_id','__group','__label']].to_csv(OUT/f'{stem}_{name}_ids.csv',index=False)
    return known,s_train,s_dev,t_support,t_query,pre
```

## Task 6. Define fixed baselines and the candidate representation
The baselines are logistic regression, random forest, a source-trained prototype network, a post-hoc target-to-source CORAL embedding alignment fitted on **unlabeled target support only**, and the candidate feature-augmented contrastive prototype model. CORALProto shares RobustProto's trained encoder and source prototypes, applying an unlabeled covariance alignment to target support and query embeddings; its source-development scores are computed without alignment. Each method sees the same source known classes, preprocessing, target query, and calibration budgets. The encoder already draws equal numbers per source class in each mini-batch, with replacement for scarce families. This addresses training-batch imbalance, but cannot correct target priors or guarantee minority-family performance. MSP/Energy use the trained linear CE head for standard post-hoc novelty scoring; that head is distinct from cosine prototype scores. Novelty is an empirical question: this candidate is an integration of existing components, and favorable results alone would not establish a new method.

```python
class Encoder(nn.Module):
    def __init__(self,d,n_classes,h=128,out=32):
        super().__init__()
        self.features=nn.Sequential(nn.Linear(d,h),nn.ReLU(),nn.Linear(h,64),nn.ReLU(),nn.Linear(64,out))
        self.head=nn.Linear(out,n_classes)
    def forward(self,x):
        z=F.normalize(self.features(x),dim=-1)
        return z,self.head(z)

def supervised_contrastive(z,y,temp=.12):
    z=F.normalize(z,dim=1)
    similarities=(z@z.T)/temp
    eye=torch.eye(len(y),device=z.device,dtype=torch.bool)
    positives=y[:,None].eq(y[None,:]) & ~eye
    similarities=similarities.masked_fill(eye,-1e9)
    logprob=similarities-torch.logsumexp(similarities,dim=1,keepdim=True)
    return -(logprob*positives).sum(dim=1)/positives.sum(dim=1).clamp_min(1).float()

def augment_features(x):
    mask=(torch.rand_like(x)>.12).float()
    return torch.clamp(x*mask + .04*torch.randn_like(x),-12,12)

def set_seed(seed):
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)
    if torch.cuda.is_available(): torch.cuda.manual_seed_all(seed)

def fit_encoder(x,y,names,seed,kind,epochs=EPOCHS):
    assert kind in {'ProtoCE','RobustProto','NoContrast','NoAugment'}
    set_seed(seed)
    model=Encoder(x.shape[1],len(names),out=CFG.embed_dim).to(DEVICE)
    optim=torch.optim.AdamW(model.parameters(),lr=CFG.lr,weight_decay=1e-4)
    perclass=[np.flatnonzero(y==i) for i in range(len(names))]
    if any(len(ii)<25 for ii in perclass): raise ValueError('At least 25 source records per class required')
    history=[]; rng=np.random.default_rng(seed)
    steps=(12 if CFG.mode=='PILOT' else 40)
    for ep in range(epochs):
        model.train(); running=[]
        for _ in range(steps):
            selections=[rng.choice(indices,size=max(8,CFG.batch_size//len(names)),replace=True) for indices in perclass]
            ids=rng.permutation(np.concatenate(selections))
            xb=torch.from_numpy(x[ids]).to(DEVICE)
            yb=torch.as_tensor(y[ids],dtype=torch.long,device=DEVICE)
            z,logits=model(xb)
            loss=F.cross_entropy(logits,yb)
            if kind!='ProtoCE':
                second=xb if kind=='NoAugment' else augment_features(xb)
                z2,_=model(second)
                if kind!='NoContrast':
                    both_z=torch.cat((z,z2)); both_y=torch.cat((yb,yb))
                    loss=loss + .06*supervised_contrastive(both_z,both_y).mean()
                if kind!='NoAugment':
                    loss=loss + .25*F.cross_entropy(model(augment_features(xb))[1],yb)
            optim.zero_grad(set_to_none=True); loss.backward(); optim.step()
            running.append(float(loss.detach().cpu()))
        history.append({'epoch':ep+1,'train_loss':float(np.mean(running))})
    model.eval()
    base=embeddings(model,x)
    prototypes=np.stack([base[y==i].mean(axis=0) for i in range(len(names))])
    prototypes=prototypes/np.maximum(np.linalg.norm(prototypes,axis=1,keepdims=True),1e-8)
    own_dist=1-np.sum(base*prototypes[y],axis=1)
    scales=np.array([max(.05,float(np.median(own_dist[y==i]))) for i in range(len(names))])
    return {'kind':kind,'model':model,'names':list(names),'prototypes':prototypes.astype('float32'),
            'scales':scales.astype('float32'),'history':history}

def fit_episodic_proto(x,y,names,seed,epochs=EPOCHS):
    # Genuine support/query episodes on SOURCE classes; target families remain unseen.
    set_seed(seed);rng=np.random.default_rng(seed)
    model=Encoder(x.shape[1],len(names),out=CFG.embed_dim).to(DEVICE)
    optim=torch.optim.AdamW(model.parameters(),lr=CFG.lr,weight_decay=1e-4)
    classes=[np.flatnonzero(y==i) for i in range(len(names))]
    history=[]
    for epoch in range(epochs):
        losses=[]
        for _ in range(12 if CFG.mode=='PILOT' else 40):
            ways=min(3,len(names)); selected=rng.choice(len(names),ways,replace=False)
            ids=[rng.choice(classes[i],size=13,replace=len(classes[i])<13) for i in selected]
            support=np.concatenate([v[:5] for v in ids]);query=np.concatenate([v[5:] for v in ids])
            z_s,_=model(torch.from_numpy(x[support]).to(DEVICE))
            z_q,_=model(torch.from_numpy(x[query]).to(DEVICE))
            centers=F.normalize(z_s.reshape(ways,5,-1).mean(1),dim=-1)
            logits=-(1-z_q@centers.T)*10
            target=torch.arange(ways,device=DEVICE).repeat_interleave(8)
            loss=F.cross_entropy(logits,target)
            optim.zero_grad(set_to_none=True);loss.backward();optim.step()
            losses.append(float(loss.detach().cpu()))
        history.append({'epoch':epoch+1,'train_loss':float(np.mean(losses))})
    base=embeddings(model,x)
    centers=np.stack([base[y==i].mean(0) for i in range(len(names))])
    centers/=np.maximum(np.linalg.norm(centers,axis=1,keepdims=True),1e-8)
    return {'kind':'EpisodicProto','model':model,'names':list(names),'prototypes':centers.astype('float32'),
            'scales':np.ones(len(names),dtype='float32'),'history':history}

@torch.no_grad()
def embeddings(model,x):
    zs=[]
    model.eval()
    for j in range(0,len(x),4096):
        z,_=model(torch.from_numpy(x[j:j+4096]).to(DEVICE))
        zs.append(z.cpu().numpy())
    return np.concatenate(zs).astype('float32')

def fit_baselines(x,y,names,seed,target_unlabeled_x):
    set_seed(seed)
    lr=None;lr_audit=[]
    for solver in ('lbfgs','saga'):
        candidate=Pipeline([('standardize_source',StandardScaler()),
            ('logreg',LogisticRegression(max_iter=1000,solver=solver,tol=1e-4,
                                       class_weight='balanced',random_state=seed))])
        with warnings.catch_warnings(record=True) as recorded:
            warnings.simplefilter('always',ConvergenceWarning)
            candidate.fit(x,y)
        trained=candidate.named_steps['logreg']
        nonconvergence=[str(w.message) for w in recorded
                        if issubclass(w.category,ConvergenceWarning)]
        converged=bool(np.max(trained.n_iter_)<trained.max_iter and not nonconvergence)
        lr_audit.append({'solver':solver,'max_iter':trained.max_iter,
                         'iterations_by_class':json.dumps(trained.n_iter_.tolist()),
                         'max_iterations_used':int(np.max(trained.n_iter_)),
                         'converged':converged,'warning':'; '.join(nonconvergence)})
        if converged:
            lr=candidate
            break
    # Failed fits stay in the audit but never enter performance comparisons.
    rf=RandomForestClassifier(n_estimators=(80 if CFG.mode=='PILOT' else 200),
                              class_weight='balanced_subsample',min_samples_leaf=2,
                              n_jobs=-1,random_state=seed)
    rf.fit(x,y)
    proto=fit_encoder(x,y,names,seed,'ProtoCE')
    # OpenMax-style activation-vector EVT, calibrated on SOURCE correct predictions.
    openmax=fit_openmax(proto,x,y)
    robust=fit_encoder(x,y,names,seed,'RobustProto')
    coral=make_coral_baseline(robust,x,target_unlabeled_x)
    packs={'RandomForest':{'kind':'RandomForest','model':rf,'names':list(names)},
            'ProtoCE':proto,
            'EpisodicProto':fit_episodic_proto(x,y,names,seed),
            'MSP':{'kind':'MSP','model':proto['model'],'names':list(names)},
            'Energy':{'kind':'Energy','model':proto['model'],'names':list(names)},
            'OpenMaxVariant':openmax,
            'RobustProto':robust,'CORALProto':coral}
    if lr is not None: packs['LogReg']={'kind':'LogReg','model':lr,'names':list(names)}
    return packs,lr_audit

def make_coral_baseline(source_pack,source_x,target_unlabeled_x,eps=1e-3):
    # Unlabeled target SUPPORT only. Map target embeddings into the source
    # geometry so the source prototypes remain fixed (target -> source CORAL).
    zs=embeddings(source_pack['model'],source_x)
    zt=embeddings(source_pack['model'],target_unlabeled_x)
    def matrix_power(cov,power):
        values,vectors=np.linalg.eigh(cov)
        return (vectors*(np.maximum(values,eps)**power))@vectors.T
    cs=np.cov(zs,rowvar=False)+eps*np.eye(zs.shape[1])
    ct=np.cov(zt,rowvar=False)+eps*np.eye(zt.shape[1])
    a=matrix_power(ct,-.5)@matrix_power(cs,.5)
    pack={k:v for k,v in source_pack.items() if k!='history'}
    pack.update({'kind':'CORALProto','mean_source':zs.mean(0),'mean_target':zt.mean(0),
                 'target_to_source':a.astype('float32'),'unlabeled_support_n':len(zt)})
    return pack

def align_coral(pack,z):
    mapped=(z-pack['mean_target'])@pack['target_to_source']+pack['mean_source']
    return mapped/np.maximum(np.linalg.norm(mapped,axis=1,keepdims=True),1e-8)

@torch.no_grad()
def class_logits(model,x):
    outputs=[];model.eval()
    for j in range(0,len(x),4096):
        _,logits=model(torch.from_numpy(x[j:j+4096]).to(DEVICE))
        outputs.append(logits.cpu().numpy())
    return np.concatenate(outputs)

def fit_openmax(proto,x,y,tail_size=20):
    from scipy.stats import weibull_min
    logits=class_logits(proto['model'],x)
    centers=[];tails=[];fit_audit=[]
    for i in range(len(proto['names'])):
        correct=logits[(y==i)&(logits.argmax(1)==i)]
        n_correct=len(correct)
        fallback=n_correct<30
        if fallback: correct=logits[y==i]
        mav=correct.mean(0);distance=np.linalg.norm(correct-mav,axis=1)
        tail=np.sort(distance)[-min(tail_size,len(distance)):]
        if np.std(tail)<1e-6:
            shape,scale=1.,max(float(np.mean(tail)),1e-6)
        else:
            try: shape,_,scale=weibull_min.fit(np.maximum(tail,1e-8),floc=0)
            except (ValueError,FloatingPointError): shape,scale=1.,max(float(np.mean(tail)),1e-6)
        centers.append(mav);tails.append((float(shape),float(scale)))
        fit_audit.append({'class':proto['names'][i],'n_training':int((y==i).sum()),
                          'n_correct':int(n_correct),'fallback_to_all_class_logits':bool(fallback),
                          'n_evt_fit':len(correct),'n_tail':len(tail)})
    return {'kind':'OpenMaxVariant','model':proto['model'],'names':list(proto['names']),
            'mavs':np.stack(centers),'evt_tails':tails,'tail_size':tail_size,
            'fit_audit':fit_audit,
            'deviation':'EVT on logit MAV; revised positive logits and augmented softmax; not canonical penultimate-activation OpenMax'}

def raw_predict(pack,x,domain='target'):
    if pack['kind'] in {'LogReg','RandomForest'}:
        prob=pack['model'].predict_proba(x)
        j=np.argmax(prob,axis=1)
        return np.asarray(pack['names'],dtype=object)[j],(1-prob[np.arange(len(j)),j]).astype('float32')
    if pack['kind'] in {'MSP','Energy','OpenMaxVariant'}:
        from scipy.special import softmax,logsumexp
        logits=class_logits(pack['model'],x)
        j=logits.argmax(1)
        if pack['kind']=='MSP': score=1-softmax(logits,axis=1).max(axis=1)
        elif pack['kind']=='Energy': score=-logsumexp(logits,axis=1)
        else:
            from scipy.stats import weibull_min
            revised=logits.copy();unknown=np.zeros(len(logits))
            top=np.argsort(-logits,axis=1)[:,:min(3,logits.shape[1])]
            for i,(shape,scale) in enumerate(pack['evt_tails']):
                mask=(top==i).any(axis=1)
                distance=np.linalg.norm(logits[mask]-pack['mavs'][i],axis=1)
                transferred=np.maximum(revised[mask,i],0)*weibull_min.cdf(distance,shape,scale=scale)
                revised[mask,i]-=transferred;unknown[mask]+=transferred
            joint=softmax(np.column_stack((revised,unknown)),axis=1)
            j=joint[:,:-1].argmax(1);score=joint[:,-1]
        return np.asarray(pack['names'],dtype=object)[j],np.asarray(score,dtype='float32')
    z=embeddings(pack['model'],x)
    if pack['kind']=='CORALProto' and domain=='target': z=align_coral(pack,z)
    dist=1-(z@pack['prototypes'].T)
    if pack['kind'] not in {'ProtoCE','EpisodicProto'}: dist=dist/pack['scales'][None,:]
    j=np.argmin(dist,axis=1)
    return np.asarray(pack['names'],dtype=object)[j],dist[np.arange(len(j)),j].astype('float32')

def with_new_class(pack,novel_name,support_x):
    if pack['kind'] not in {'ProtoCE','RobustProto','EpisodicProto','CORALProto'}:
        return None  # Frozen classical baselines only; no false claim of incremental retraining.
    new=dict(pack)
    z=embeddings(pack['model'],support_x)
    if pack['kind']=='CORALProto': z=align_coral(pack,z)
    center=z.mean(axis=0);center=center/max(float(np.linalg.norm(center)),1e-8)
    new['prototypes']=np.vstack([pack['prototypes'],center.astype('float32')])
    new['scales']=np.append(pack['scales'],float(np.median(pack['scales'])))
    new['names']=list(pack['names'])+[novel_name]
    return new

def prototype_scores_from_z(pack,z_query,z_support,novel_name):
    center=z_support.mean(0);center/=max(float(np.linalg.norm(center)),1e-8)
    prototypes=np.vstack((pack['prototypes'],center))
    dist=1-z_query@prototypes.T
    if pack['kind'] in {'RobustProto','CORALProto'}:
        scale=np.r_[pack['scales'],float(np.median(pack['scales']))]
        dist=dist/scale[None,:]
    j=dist.argmin(1);names=np.asarray(pack['names']+[novel_name],dtype=object)
    return names[j],dist[np.arange(len(j)),j]

def calibrate_threshold(pack,dev_x,quantile):
    _, scores=raw_predict(pack,dev_x,domain='source')
    return float(np.quantile(scores,quantile))

def reject_predict(pack,x,threshold):
    pred,score=raw_predict(pack,x)
    pred=pred.copy();pred[score>threshold]='Unknown'
    return pred,score
```

## Task 7. Metrics and calibrated operating points
Classification and unknown-rejection metrics are separate. `benign_alert_rate` counts any benign prediction other than Benign. AUROC/AUPR refer to unknown-family detection, not multiclass accuracy. The pooled unknown AUROC and the family-wise macro unknown AUROC use different weighting. Each method's score and threshold have their own scale; numerical threshold values are not compared across methods. OSCR integrates known correct acceptance against unknown false acceptance over thresholds. Source, unlabeled target support, and labeled target-known support are distinct calibration protocols; only the post-hoc oracle uses query labels, strictly for diagnosis.

```python
def oscr_score(truth,pred_closed,unknown_score,known):
    truth=np.asarray(truth,dtype=object); pred_closed=np.asarray(pred_closed,dtype=object)
    known_mask=np.isin(truth,known); unk_mask=~known_mask
    if known_mask.sum()==0 or unk_mask.sum()==0: return np.nan
    order=np.argsort(unknown_score,kind='stable')
    scores=np.asarray(unknown_score)[order]
    correct=(known_mask & (pred_closed==truth))[order]
    unk=unk_mask[order]
    end=np.r_[np.flatnonzero(np.diff(scores)),len(scores)-1]
    ccr=np.r_[0,np.cumsum(correct)[end]/known_mask.sum()]
    fpr=np.r_[0,np.cumsum(unk)[end]/unk_mask.sum()]
    return float(scipy.integrate.trapezoid(ccr,fpr))

def score_predictions(truth,pred,unknown_score,known,closed_pred=None):
    truth=np.asarray(truth,dtype=object);pred=np.asarray(pred,dtype=object)
    is_unk=~np.isin(truth,known);flag=pred=='Unknown'; benign=truth=='Benign'
    valid_known=~is_unk
    d={'n_query':len(truth),'n_unknown':int(is_unk.sum()),'n_benign':int(benign.sum()),
       'known_macro_f1':float(f1_score(truth[valid_known],pred[valid_known],labels=known,average='macro',zero_division=0)),
       'unknown_recall':float(flag[is_unk].mean()) if is_unk.any() else np.nan,
       'unknown_precision':float(is_unk[flag].mean()) if flag.any() else 0.,
       'benign_rejection_rate':float(flag[benign].mean()) if benign.any() else np.nan,
       'benign_accepted_misclassification_rate':float(
           ((pred[benign]!='Benign') & (pred[benign]!='Unknown')).mean()) if benign.any() else np.nan,
       'benign_alert_rate':float((pred[benign]!='Benign').mean()) if benign.any() else np.nan,
       'known_rejection_rate':float(flag[valid_known].mean()) if valid_known.any() else np.nan}
    # A second exact partition identifies the floor no rejection threshold
    # can remove while the closed-set classifier itself stays fixed. For a
    # benign flow misclassified as a known attack before thresholding, either
    # that attack prediction or a subsequent Unknown rejection is an alert.
    if closed_pred is not None and benign.any():
        closed=np.asarray(closed_pred,dtype=object)
        d['benign_closed_error_floor']=float((closed[benign]!='Benign').mean())
        d['benign_correct_closed_rejection_rate']=float(
            (flag[benign] & (closed[benign]=='Benign')).mean())
        assert np.isclose(d['benign_alert_rate'],
                          d['benign_closed_error_floor']+
                          d['benign_correct_closed_rejection_rate'])
    else:
        d['benign_closed_error_floor']=np.nan
        d['benign_correct_closed_rejection_rate']=np.nan
    families=sorted(set(truth[is_unk]))
    d['unknown_macro_recall']=float(np.mean([np.mean(flag[truth==family]) for family in families])) if families else np.nan
    if is_unk.any() and valid_known.any():
        d['unknown_auroc']=float(roc_auc_score(is_unk,unknown_score))
        d['unknown_aupr']=float(average_precision_score(is_unk,unknown_score))
        d['unknown_macro_auroc']=float(np.mean([
            roc_auc_score(np.r_[np.zeros(valid_known.sum()),np.ones((truth==family).sum())],
                            np.r_[unknown_score[valid_known],unknown_score[truth==family]])
            for family in families]))
        if closed_pred is not None: d['oscr']=oscr_score(truth,closed_pred,unknown_score,known)
    return d

def unavailable_geometry(reason,n_centroids=0):
    return {'pseudo_class_prototype_cosine_shift':np.nan,
            'n_pseudo_classes_with_centroids':n_centroids,
            'geometry_reference':'shared source-trained ProtoCE',
            'geometry_unavailable':True,
            'geometry_unavailable_reason':reason}

def protoce_pair_geometry(protoce_pack,x_support):
    # One shared source-trained geometry measurement per pair/seed. The
    # target-support rows are unlabeled; pseudo-labels come from ProtoCE.
    assert protoce_pack['kind']=='ProtoCE' and 'prototypes' in protoce_pack
    predicted,_=raw_predict(protoce_pack,x_support)
    embedded=embeddings(protoce_pack['model'],x_support)
    distances=[]
    for i,name in enumerate(protoce_pack['names']):
        selected=predicted==name
        if selected.sum()<10: continue
        centroid=embedded[selected].mean(0)
        centroid/=max(float(np.linalg.norm(centroid)),1e-8)
        distances.append(float(1-np.clip(centroid@protoce_pack['prototypes'][i],-1,1)))
    if not distances:
        return unavailable_geometry('Fewer than 10 target-support rows for every ProtoCE pseudo-class')
    value=float(np.mean(distances))
    if not np.isfinite(value):
        return unavailable_geometry('Nonfinite pseudo-class centroid/prototype distance',len(distances))
    return {'pseudo_class_prototype_cosine_shift':value,
            'n_pseudo_classes_with_centroids':len(distances),
            'geometry_reference':'shared source-trained ProtoCE',
            'geometry_unavailable':False,'geometry_unavailable_reason':''}

def prequery_shift_indicators(source_closed,source_scores,
                               target_closed,target_scores,pair_geometry):
    # The three score indicators are method specific. Geometry is computed
    # once using ProtoCE. No target-support label or target-query value enters.
    from scipy.stats import wasserstein_distance
    sd=np.asarray(source_scores,dtype=float);ts=np.asarray(target_scores,dtype=float)
    scale=max(float(np.subtract(*np.percentile(sd,[75,25]))),1e-8)
    s=(sd-float(np.median(sd)))/scale
    t=(ts-float(np.median(sd)))/scale
    dev_b=source_closed=='Benign';target_b=target_closed=='Benign'
    benign_w1=(float(wasserstein_distance(s[dev_b],t[target_b]))
               if dev_b.sum()>=20 and target_b.sum()>=20 else np.nan)
    dev_cut=float(np.quantile(sd,.95))
    return {'score_wasserstein_source_iqr_units':float(wasserstein_distance(s,t)),
            'predicted_benign_score_wasserstein_source_iqr_units':benign_w1,
            'n_source_predicted_benign':int(dev_b.sum()),
            'n_target_predicted_benign':int(target_b.sum()),
            **pair_geometry,
            'source_tail_exceedance_delta':float(np.mean(ts>dev_cut)-np.mean(sd>dev_cut)),
            'no_query_labels_in_predictors':True}

def evaluate(pack,threshold,pre,target,known):
    x=pre.transform(target)
    truth=target['__label'].to_numpy(dtype=object)
    closed,_=raw_predict(pack,x)
    pred,unknown_score=reject_predict(pack,x,threshold)
    return score_predictions(truth,pred,unknown_score,known,closed),pred,unknown_score

def save_predictions(path,frame,pred,score,stage,novel=None):
    out=frame[['__dataset','__row_id','__group','__label']].copy()
    out['stage']=stage;out['adapted_family']=novel or ''
    out['prediction']=pred;out['unknown_score']=score
    out.to_parquet(path,index=False,compression='zstd')

def save_confusion(path,truth,pred,known,title):
    actual=np.where(np.isin(truth,known),truth,'Unknown')
    labels=list(known)+['Unknown']
    m=confusion_matrix(actual,pred,labels=labels)
    pd.DataFrame(m,index=labels,columns=labels).to_csv(str(path).replace('.png','.csv'))
    plt.figure(figsize=(max(6,len(labels)*.9),max(5,len(labels)*.75)))
    sns.heatmap(m,annot=True,fmt='d',xticklabels=labels,yticklabels=labels,cmap='Blues')
    plt.xlabel('Prediction');plt.ylabel('Ground truth');plt.title(title);plt.tight_layout()
    plt.savefig(path,dpi=180);plt.close()

def select_target_known_support(target_known_support,known,budget,seed):
    selections=[]
    for i,name in enumerate(known):
        pool=target_known_support[target_known_support['__label'].eq(name)]
        count=len(pool) if budget=='all' else budget
        if len(pool)<count: raise ValueError(f'{name} has {len(pool)} target calibration examples; {count} required')
        # A nested, deterministic selection makes B=5,10,20,50 comparable.
        selections.append(pool.sample(frac=1,random_state=seed+100*i).iloc[:count])
    return pd.concat(selections,ignore_index=True)

def threshold_from_known_scores(scores,labels,q,rule='max_guard'):
    scores=np.asarray(scores,dtype=float)
    benign=np.asarray(labels,dtype=object)=='Benign'
    if not benign.any() or not len(scores) or not np.isfinite(scores).all():
        raise ValueError('Threshold requires finite known-class scores including Benign')
    pooled=float(np.quantile(scores,q))
    benign_q=float(np.quantile(scores[benign],q))
    benign_max=float(np.max(scores[benign]))
    if rule=='pooled_q': return pooled
    if rule=='benign_q': return benign_q
    if rule=='max_guard': return max(pooled,benign_max)
    raise ValueError(f'Unknown threshold rule: {rule}')

def target_threshold(pack,pre,target_known_support,known,budget,seed,q,rule='max_guard'):
    calibration=select_target_known_support(target_known_support,known,budget,seed)
    # A distinct, labeled target-known budget; no target query labels and no novel-family support.
    _,scores=raw_predict(pack,pre.transform(calibration))
    # Finite-sample benign guard: max of B benign calibration scores has marginal
    # rejection probability <=1/(B+1) for exchangeable target benign flows. B=20
    # permits ~4.76%; B=5/10 are sensitivity analyses with weaker bounds.
    # Benign misclassification is unaffected. The pooled and benign-only quantile
    # rules provide an explicit ablation; none uses target query labels.
    threshold=threshold_from_known_scores(scores,calibration['__label'],q,rule)
    return threshold,calibration

def unlabeled_gmm_threshold(pack,pre,unlabeled_pool,seed,q,source_cut,precomputed_scores=None):
    # Transductive calibration from a disjoint target support pool with labels hidden.
    # The lower-score component is a heuristic for target known traffic; the mixture
    # can fail when source scores do not order known/unknown or the sampled priors shift.
    if precomputed_scores is None:
        _,scores=raw_predict(pack,pre.transform(unlabeled_pool.drop(columns=['__label'],errors='ignore')))
    else:
        scores=precomputed_scores  # Source-only preprocessor ignores labels by construction.
    scores=np.asarray(scores,dtype=float)
    if len(scores)<60 or not np.isfinite(scores).all():
        return source_cut,{'status':'source_fallback_invalid_scores','n_unlabeled':len(scores)}
    lo,hi=np.quantile(scores,[.01,.99]);clipped=np.clip(scores,lo,hi).reshape(-1,1)
    if np.std(clipped)<1e-8:
        return source_cut,{'status':'source_fallback_constant_scores','n_unlabeled':len(scores)}
    try:
        gm=GaussianMixture(n_components=2,covariance_type='full',reg_covar=1e-5,
                           n_init=3,max_iter=200,random_state=seed).fit(clipped)
    except (ValueError,np.linalg.LinAlgError) as err:
        return source_cut,{'status':'source_fallback_fit_error','n_unlabeled':len(scores),
                           'reason':str(err)}
    lower=int(np.argmin(gm.means_.ravel()))
    selected=gm.predict_proba(clipped)[:,lower]>=.7
    fraction=float(selected.mean())
    if selected.sum()<20 or fraction<.1 or fraction>.9:
        return source_cut,{'status':'source_fallback_unseparated_mixture',
                           'n_unlabeled':len(scores),'lower_component_fraction':fraction}
    threshold=float(np.quantile(scores[selected],q))
    return threshold,{'status':'gmm_lower_component','n_unlabeled':len(scores),
                      'lower_component_fraction':fraction,
                      'component_means':gm.means_.ravel().tolist()}

def open_set_utility(truth,closed,score,known,threshold):
    truth=np.asarray(truth,dtype=object);known_mask=np.isin(truth,known);unknown=~known_mask
    if not known_mask.any() or not unknown.any(): return np.nan
    correct=np.mean((closed[known_mask]==truth[known_mask]) & (score[known_mask]<=threshold))
    recall=np.mean(score[unknown]>threshold)
    return float(.5*(correct+recall))

def query_oracle_diagnostic(truth,closed,score,known):
    # Non-deployable post hoc ceiling. Never pass the chosen cut to evaluate().
    truth=np.asarray(truth,dtype=object);known_mask=np.isin(truth,known);unknown=~known_mask
    if not known_mask.any() or not unknown.any(): return {'oracle_threshold':np.nan,'oracle_utility':np.nan}
    order=np.argsort(score,kind='stable');sorted_score=np.asarray(score)[order]
    correct=(known_mask & (closed==truth))[order];unk=unknown[order]
    ends=np.r_[np.flatnonzero(np.diff(sorted_score)),len(score)-1]
    cuts=np.r_[-np.inf,sorted_score[ends]]
    accepted=np.r_[0,np.cumsum(correct)[ends]]/known_mask.sum()
    unknown_rejected=1-np.r_[0,np.cumsum(unk)[ends]]/unknown.sum()
    values=.5*(accepted+unknown_rejected)
    best=int(np.argmax(values))
    return {'oracle_threshold':float(cuts[best]),'oracle_utility':float(values[best])}

def family_rows(truth,pred,score,known,base):
    known_mask=np.isin(truth,known)
    out=[]
    for family in sorted(set(truth)-set(known)):
        mask=truth==family
        out.append({**base,'family':family,'n':int(mask.sum()),
                    'unknown_recall':float(np.mean(pred[mask]=='Unknown')),
                    'auroc_vs_known':float(roc_auc_score(
                        np.r_[np.zeros(known_mask.sum()),np.ones(mask.sum())],
                        np.r_[score[known_mask],score[mask]])) if known_mask.any() else np.nan})
    return out

def measure_fewshot_update(pack,novel_name,support_x,repetitions=100):
    # Gross GPU board energy, if the driver exposes an energy counter; no fabricated CO2.
    sync=lambda:torch.cuda.synchronize() if DEVICE.type=='cuda' else None
    handle=None;nvml=None;energy_start=None
    base_mem=np.nan
    if DEVICE.type=='cuda':
        try:
            import pynvml
            pynvml.nvmlInit();handle=pynvml.nvmlDeviceGetHandleByIndex(torch.cuda.current_device())
            energy_start=pynvml.nvmlDeviceGetTotalEnergyConsumption(handle)
            nvml=pynvml
        except Exception: pass
        torch.cuda.reset_peak_memory_stats()
        base_mem=torch.cuda.memory_allocated()
    sync();tic=time.perf_counter()
    for _ in range(repetitions):
        _=with_new_class(pack,novel_name,support_x)
    sync();elapsed=time.perf_counter()-tic
    try: energy_mj=(nvml.nvmlDeviceGetTotalEnergyConsumption(handle)-energy_start)/repetitions if nvml else np.nan
    except Exception: energy_mj=np.nan
    return {'update_wall_ms':elapsed*1000/repetitions,
            'gross_gpu_board_energy_mj_per_update':energy_mj,
            'gpu_incremental_peak_bytes':max(0,torch.cuda.max_memory_allocated()-base_mem) if DEVICE.type=='cuda' else np.nan,
            'energy_note':'Gross board counter includes idle draw; NA if unsupported; includes support encoding and prototype update, excludes precomputed query embedding'}

@torch.no_grad()
def measure_batched_updates(pack,pool_x,shots,seed,n_sets=1000,batch_sets=128,repetitions=3):
    # Batch distinct within-set support draws. Inputs are preprocessed; timing
    # includes GPU/CPU support encoding and prototype assembly, excludes I/O.
    if len(pool_x)<shots: raise ValueError('Insufficient support pool for batched benchmark')
    rng=np.random.default_rng(seed)
    support_ids=np.stack([rng.choice(len(pool_x),size=shots,replace=False) for _ in range(n_sets)])
    support=pool_x[support_ids]
    sync=lambda:torch.cuda.synchronize() if DEVICE.type=='cuda' else None
    measurements=[]
    for _ in range(repetitions):
        sync();tic=time.perf_counter()
        for start in range(0,n_sets,batch_sets):
            part=support[start:start+batch_sets]
            z=embeddings(pack['model'],part.reshape(-1,pool_x.shape[1]))
            if pack['kind']=='CORALProto': z=align_coral(pack,z)
            z=z.reshape(len(part),shots,-1)
            center=z.mean(1)
            center/=np.maximum(np.linalg.norm(center,axis=1,keepdims=True),1e-8)
            _=np.concatenate((np.broadcast_to(pack['prototypes'],
                   (len(center),*pack['prototypes'].shape)),center[:,None,:]),axis=1)
            _=np.concatenate((np.broadcast_to(pack['scales'],
                   (len(center),len(pack['scales']))),
                   np.full((len(center),1),float(np.median(pack['scales'])))),axis=1)
        sync();measurements.append(time.perf_counter()-tic)
    return {'measurement':'batched_sets','n_support_sets':n_sets,'batch_sets':batch_sets,
            'throughput_sets_per_second':n_sets/float(np.median(measurements)),
            'median_wall_ms_per_set':1000*float(np.median(measurements))/n_sets,
            'timing_scope':'support encoding and prototype assembly; preprocessed input; excludes data I/O and query scoring'}
```

```python
def protocol_self_checks():
    mapping={'SSH-Bruteforce':'BruteForce','FTP-BruteForce':'BruteForce',
             'Brute Force -Web':'Web Attacks','SQL Injection':'Web Attacks',
             'DDoS attacks-LOIC-HTTP':'DDoS','DoS attacks-Slowloris':'DoS',
             'Infilteration':'Infiltration'}
    assert all(canonical_label(raw,'CIC')==expected for raw,expected in mapping.items())
    synthetic_scores=np.array([.1,.25,.3,.45,.8])
    synthetic_labels=np.array(['Benign','DoS','Benign','DoS','DoS'])
    assert threshold_from_known_scores(synthetic_scores,synthetic_labels,.8,'max_guard') >= (
           threshold_from_known_scores(synthetic_scores,synthetic_labels,.8,'pooled_q'))
    assert threshold_from_known_scores(synthetic_scores,synthetic_labels,.8,'max_guard') >= (
           threshold_from_known_scores(synthetic_scores,synthetic_labels,.8,'benign_q'))
    truth=np.array(['Benign','DoS','Benign','Novel','Novel'],dtype=object)
    closed=np.array(['Benign','DoS','DoS','DoS','Benign'],dtype=object)
    scores=np.array([.10,.15,.25,.50,.75])
    known=['Benign','DoS'];oracle=query_oracle_diagnostic(truth,closed,scores,known)
    assert 0<=oscr_score(truth,closed,scores,known)<=1
    for cut in (.1,.3,.6):
        assert oracle['oracle_utility']+1e-12>=open_set_utility(truth,closed,scores,known,cut)
    cuts=[.1,.3,.6]
    rejected=[np.mean(scores[truth=='Benign']>v) for v in cuts]
    assert rejected==sorted(rejected,reverse=True)
    pred=closed.copy();pred[scores>.3]='Unknown'
    got=score_predictions(truth,pred,scores,known,closed)
    assert np.isclose(got['benign_rejection_rate'],.0)
    assert np.isclose(got['benign_alert_rate'],.5)  # Closed-set error survives threshold calibration.
    # Exact repeated vectors stay indivisible; quarantine removes conflicting labels.
    global COMMON_FEATURES
    original=COMMON_FEATURES
    try:
        COMMON_FEATURES=['SYNTHETIC_FEATURE']
        frame=pd.DataFrame({'SYNTHETIC_FEATURE':[1,1,2],
                            '__label':['Benign','DoS','Benign']})
        quarantine,report=clean_and_group(frame,'synthetic')
        retained,_=clean_and_group(frame,'synthetic',policy='retain_grouped')
        assert report['conflicting_rows_removed']==2 and len(quarantine)==1
        assert retained.iloc[0].__group==retained.iloc[1].__group
    finally: COMMON_FEATURES=original
    print('Synthetic protocol checks passed')

protocol_self_checks()
```

## Task 8. Cross-dataset, calibrated, episodic experiments
The **source** threshold uses only source development data. The **target-unlabeled-GMM** threshold fits a two-component score mixture to the disjoint target support pool after hiding all its labels; this heuristic has no false-alert guarantee and may be distorted by class-capped sampling. The **target-known** main threshold uses B=20 labeled support records from *each known family* and is a separately disclosed label budget. The rule/budget study also compares B=0, 5, 10, 20, 50 and all available target-known support with pooled quantile, benign quantile, and a max guard. The max guard is the larger of the pooled known quantile and the largest sampled benign score; under exchangeable benign observations it bounds benign rejection at 1/(B+1), not all benign alerts. No deployable threshold uses target query labels. An oracle using query labels is computed afterward for diagnostic headroom only and stored in a separate file. Independent support draws, not nested draws, define each novel-family episode; the separate budget study uses nested support draws across B. Thresholds share one trained model and exactly the same target query. The PILOT maximizes distinct source/target coverage and records infeasible shortfalls; FINAL runs every eligible pair. Checkpoints prevent interrupted long runs from repeating completed pair–seed jobs.

```python
eligible=PAIRS[PAIRS.eligible].sort_values(['source','target']).reset_index(drop=True)
if CFG.mode=='PILOT':
    preferred=[('BOT','NB15'),('CIC','BOT'),('NB15','TON'),('TON','CIC')]
    from itertools import combinations
    options=list(eligible.itertuples(index=False))
    # Maximise distinct source and target coverage. Preferred pairs break ties;
    # a reduced pilot is reported explicitly if four-way coverage is infeasible.
    candidates=[]
    for count in range(min(CFG.pilot_pair_count,len(options)),0,-1):
        candidates=[set_ids for set_ids in combinations(range(len(options)),count)
            if len({options[i].source for i in set_ids})==count
            and len({options[i].target for i in set_ids})==count]
        if candidates: break
    if not candidates: raise ValueError('No domain-distinct pilot pairs are eligible')
    choice=max(candidates,key=lambda ids:sum((options[i].source,options[i].target) in preferred for i in ids))
    eligible=eligible.iloc[list(choice)].sort_values(['source','target']).reset_index(drop=True)
    assert eligible.source.is_unique and eligible.target.is_unique
    if len(eligible)<CFG.pilot_pair_count:
        warnings.warn(f'Only {len(eligible)} domain-distinct pairs are eligible for a {CFG.pilot_pair_count}-pair pilot')
    (OUT/'pilot_pair_selection.json').write_text(json.dumps({
        'preferred_pairs':preferred,'requested':CFG.pilot_pair_count,'selected':eligible[['source','target']].to_dict('records'),
        'distinct_sources':eligible.source.is_unique,'distinct_targets':eligible.target.is_unique,
        'eligible_total':len(options)},indent=2))
print('Eligible directed pairs:',int(PAIRS.eligible.sum()),'| Executing:',eligible[['source','target']].to_dict('records'))
if not (eligible.source.eq('CIC')|eligible.target.eq('CIC')).any():
    warnings.warn('CIC is still absent from experiments: inspect label_harmonization.csv')
RESULTS=[];FAMILY=[];TRAIN_HISTORY=[];EPISODE_SUPPORT=[];COMPUTE=[];ORACLE=[];CALIBRATION_AUDIT=[]
RULE_BUDGET=[];OPENMAX_AUDIT=[];LOGREG_AUDIT=[];REPRESENTATION=[];PREDICTORS=[]
first_example=None
config_hash=hashlib.sha256(json.dumps({'config':asdict(CFG),'inputs':provenance['inputs'],
     'protocol':'v8-reviewer-geometry-bootstrap-coverage-20260930'},sort_keys=True,default=str).encode()).hexdigest()
anchor_key=(eligible.iloc[0].source,eligible.iloc[0].target,SEEDS[0])

def episode_indices(src,tgt,seed,novel,k,episode,pool_size):
    key=f'{src}|{tgt}|{seed}|{novel}|{k}|{episode}'.encode()
    episode_seed=int.from_bytes(hashlib.sha256(key).digest()[:8],'little')
    return np.random.default_rng(episode_seed).choice(pool_size,size=k,replace=False)

for pair in eligible.itertuples(index=False):
  for seed in SEEDS:
    src,tgt=pair.source,pair.target
    run_id=f'{src}_to_{tgt}_seed{seed}'
    checkpoint=OUT/f'{run_id}_reviewer_completed.json'
    if checkpoint.exists() and (src,tgt,seed)!=anchor_key:
        saved=json.loads(checkpoint.read_text())
        if saved.get('config_hash')==config_hash:
            for key,buffer in [('results',RESULTS),('family',FAMILY),('history',TRAIN_HISTORY),
                               ('support',EPISODE_SUPPORT),('compute',COMPUTE),
                               ('oracle',ORACLE),('calibration_audit',CALIBRATION_AUDIT),
                               ('rule_budget',RULE_BUDGET),('openmax_audit',OPENMAX_AUDIT),
                               ('logreg_audit',LOGREG_AUDIT),('representation',REPRESENTATION),
                               ('predictors',PREDICTORS)]: buffer.extend(saved[key])
            print('Reused:',run_id);continue
    starts=[len(a) for a in (RESULTS,FAMILY,TRAIN_HISTORY,EPISODE_SUPPORT,COMPUTE,
                            ORACLE,CALIBRATION_AUDIT,RULE_BUDGET,OPENMAX_AUDIT,
                            LOGREG_AUDIT,REPRESENTATION,PREDICTORS)]
    tic=time.perf_counter();print('Running:',run_id,'episodes:',EPISODES)
    known,s_train,s_dev,t_support,t_query,pre=prepare_pair(src,tgt,seed)
    assert set(t_support.__group).isdisjoint(set(t_query.__group))
    y=s_train['__label'].map({v:i for i,v in enumerate(known)}).to_numpy(dtype=np.int64)
    x_train=pre.transform(s_train);x_dev=pre.transform(s_dev);x_query=pre.transform(t_query)
    x_support=pre.transform(t_support)
    support_row_ids=pd.Index(t_support['__row_id'].to_numpy())
    assert support_row_ids.is_unique
    truth=t_query['__label'].to_numpy(dtype=object)
    packs,lr_attempts=fit_baselines(x_train,y,known,seed,x_support)
    try:
        shared_geometry=protoce_pair_geometry(packs['ProtoCE'],x_support)
    except (ValueError,FloatingPointError,np.linalg.LinAlgError) as err:
        # A failed diagnostic must not discard otherwise valid classifiers.
        # Unexpected model or programming errors still propagate.
        shared_geometry=unavailable_geometry(f'{type(err).__name__}: {err}')
    if shared_geometry['geometry_unavailable']:
        warnings.warn(f'{run_id}: ProtoCE geometry unavailable: '
                      f"{shared_geometry['geometry_unavailable_reason']}")
    LOGREG_AUDIT.extend([{'source':src,'target':tgt,'seed':seed,**attempt} for attempt in lr_attempts])
    if not any(row['converged'] for row in lr_attempts):
        print('LogReg omitted from',run_id,': neither prespecified solver converged')
    model_dir=OUT/'trained_models';model_dir.mkdir(exist_ok=True)
    for saved_method in ('ProtoCE','EpisodicProto','RobustProto','CORALProto'):
        saved_pack=packs[saved_method]
        torch.save({'state_dict':saved_pack['model'].state_dict(),'kind':saved_method,
                    'feature_names':pre.keep,'known_names':known,'prototypes':saved_pack['prototypes'],
                    'scales':saved_pack['scales'],'seed':seed,'config_hash':config_hash,
                    'coral_alignment':{k:saved_pack[k] for k in ('mean_source','mean_target','target_to_source')}
                       if saved_method=='CORALProto' else None},
                   model_dir/f'{run_id}_{saved_method}.pt')
    for method,pack in packs.items():
      if 'history' in pack:
        TRAIN_HISTORY.extend([{'source':src,'target':tgt,'seed':seed,'method':method,**h} for h in pack['history']])
      if method=='OpenMaxVariant':
          OPENMAX_AUDIT.extend([{'source':src,'target':tgt,'seed':seed,'method':method,**row}
                                for row in pack['fit_audit']])
      source_cut=calibrate_threshold(pack,x_dev,CFG.known_acceptance_quantile)
      support_closed,support_scores=raw_predict(pack,x_support)
      calibration=select_target_known_support(t_support,known,CFG.target_calibration_per_known_class,seed)
      cal_positions=support_row_ids.get_indexer(calibration['__row_id'])
      assert (cal_positions>=0).all()
      target_cut=threshold_from_known_scores(support_scores[cal_positions],calibration['__label'],
                   CFG.known_acceptance_quantile)
      assert set(calibration.__group).isdisjoint(set(t_query.__group))
      unlabeled_cut,gmm_audit=unlabeled_gmm_threshold(pack,pre,t_support,seed,
                   CFG.known_acceptance_quantile,source_cut,precomputed_scores=support_scores)
      CALIBRATION_AUDIT.append({'source':src,'target':tgt,'seed':seed,'method':method,
                   'threshold':unlabeled_cut,**gmm_audit})
      cuts={'source':source_cut,'target_unlabeled_gmm':unlabeled_cut,'target_known':target_cut}
      query_closed,query_score=raw_predict(pack,x_query)
      diagnostic=query_oracle_diagnostic(truth,query_closed,query_score,known)
      source_closed,source_scores=raw_predict(pack,x_dev,domain='source')
      source_truth=s_dev['__label'].to_numpy(dtype=object)
      known_query=np.isin(truth,known)
      assert known_query.any()
      source_f1=float(f1_score(source_truth,source_closed,labels=known,
                               average='macro',zero_division=0))
      target_f1=float(f1_score(truth[known_query],query_closed[known_query],
                               labels=known,average='macro',zero_division=0))
      REPRESENTATION.append({'source':src,'target':tgt,'seed':seed,'method':method,
          'source_dev_closed_macro_f1':source_f1,
          'target_query_known_closed_macro_f1':target_f1,
          'closed_macro_f1_change_target_minus_source':target_f1-source_f1,
          'source_dev_benign_closed_misclassification_rate':float(
              (source_closed[source_truth=='Benign']!='Benign').mean()),
          'target_query_benign_closed_misclassification_rate':float(
              (query_closed[truth=='Benign']!='Benign').mean()),
          'interpretation':'total closed-set domain-transfer gap, not isolated encoder cost'})
      PREDICTORS.append({'source':src,'target':tgt,'seed':seed,'method':method,
          **prequery_shift_indicators(source_closed,source_scores,
                                      support_closed,support_scores,shared_geometry),
          'source_threshold_oracle_gap_diagnostic_only':diagnostic['oracle_utility']-
               open_set_utility(truth,query_closed,query_score,known,source_cut)})
      # Every budget/rule uses the same fitted model and fixed query. B=0 is the
      # source-only operating point; 'all' means every known target support row.
      for budget in (0,5,10,20,50,'all'):
          if budget==0:
              budget_cuts=[('source_only',source_cut,0)]
          else:
              try: budget_cal=select_target_known_support(t_support,known,budget,seed)
              except ValueError as err:
                  RULE_BUDGET.append({'source':src,'target':tgt,'seed':seed,'method':method,
                                      'budget_per_known_class':budget,'status':'infeasible','reason':str(err)})
                  continue
              positions=support_row_ids.get_indexer(budget_cal['__row_id'])
              assert (positions>=0).all()
              budget_scores=support_scores[positions]
              budget_cuts=[(rule,threshold_from_known_scores(budget_scores,budget_cal['__label'],
                              CFG.known_acceptance_quantile,rule),len(budget_cal))
                           for rule in ('pooled_q','benign_q','max_guard')]
          for rule,cut,n_labeled in budget_cuts:
              budget_pred=query_closed.copy();budget_pred[query_score>cut]='Unknown'
              measured=score_predictions(truth,budget_pred,query_score,known,query_closed)
              utility=open_set_utility(truth,query_closed,query_score,known,cut)
              RULE_BUDGET.append({'source':src,'target':tgt,'seed':seed,'method':method,
                       'budget_per_known_class':budget,'rule':rule,'n_labeled_total':n_labeled,
                       'status':'completed','threshold':cut,'threshold_utility':utility,
                       'oracle_gap':diagnostic['oracle_utility']-utility,**measured})
      for cal,cut in cuts.items():
          utility=open_set_utility(truth,query_closed,query_score,known,cut)
          ORACLE.append({'source':src,'target':tgt,'seed':seed,'method':method,
                         'calibration':cal,'stage':'pre','threshold':cut,
                         'threshold_utility':utility,**diagnostic,
                         'oracle_gap':diagnostic['oracle_utility']-utility,
                         'diagnostic_only':'query labels used only to compute diagnostic; oracle threshold never used for prediction'})
      pre_metrics={}
      for cal,cut in cuts.items():
        stats,pred,score=evaluate(pack,cut,pre,t_query,known)
        base={'source':src,'target':tgt,'seed':seed,'method':method,'calibration':cal,
              'stage':'pre','novel_family':'','shots':0,'episode':0,'threshold':cut,
              'target_known_calibration_n':len(calibration) if cal=='target_known' else 0,
              'target_unlabeled_n':len(t_support) if cal=='target_unlabeled_gmm' else 0}
        RESULTS.append({**base,**stats});FAMILY.extend(family_rows(truth,pred,score,known,base))
        pre_metrics[cal]=stats
        if method=='RobustProto':
            save_predictions(OUT/f'{run_id}_{method}_{cal}_pre.parquet',t_query,pred,score,'pre')
        if first_example is None and method=='RobustProto' and cal=='target_known':
            first_example={'source':src,'target':tgt,'seed':seed,'known':known,
                           'novel':pair.unknown_choices[0],'pre':pre,'s_train':s_train,
                           's_dev':s_dev,'t_support':t_support,'t_query':t_query,
                           'x_train':x_train,'y_train':y,'pack':pack,'thresholds':cuts,
                           'pre_predictions':pred.copy()}
            save_confusion(OUT/'example_pre_confusion.png',truth,pred,known,
                           f'{src} to {tgt}: target-known threshold, seed {seed}')
      if method not in {'ProtoCE','RobustProto','EpisodicProto','CORALProto'}: continue
      z_query=embeddings(pack['model'],x_query)
      if method=='CORALProto': z_query=align_coral(pack,z_query)
      old_mask=np.isin(truth,known)
      pre_old_f1={cal:f1_score(truth[old_mask],
                      reject_predict(pack,x_query,cut)[0][old_mask],
                      labels=known,average='macro',zero_division=0) for cal,cut in cuts.items()}
      for novel in pair.unknown_choices:
        pool=t_support[t_support['__label'].eq(novel)]
        x_pool=pre.transform(pool);z_pool=embeddings(pack['model'],x_pool)
        if method=='CORALProto': z_pool=align_coral(pack,z_pool)
        novel_mask=truth==novel;remaining=~np.isin(truth,known+[novel])
        if seed==SEEDS[0] and novel==pair.unknown_choices[0]:
            for budget in CFG.shots:
                ids=episode_indices(src,tgt,seed,novel,budget,0,len(pool))
                COMPUTE.append({'source':src,'target':tgt,'seed':seed,'method':method,
                                'novel_family':novel,'shots':budget,'repetitions':100,
                                'measurement':'single_set_repeated',
                                **measure_fewshot_update(pack,novel,x_pool[ids])})
                if (src,tgt,seed)==anchor_key:
                    COMPUTE.append({'source':src,'target':tgt,'seed':seed,'method':method,
                        'novel_family':novel,'shots':budget,
                        **measure_batched_updates(pack,x_pool,budget,seed+budget)})
        for k in CFG.shots:
          for ep in range(EPISODES):
            ids=episode_indices(src,tgt,seed,novel,k,ep,len(pool))
            if method=='ProtoCE':
                EPISODE_SUPPORT.append({'source':src,'target':tgt,'seed':seed,'novel_family':novel,
                   'shots':k,'episode':ep,'support_row_ids':json.dumps(pool.iloc[ids]['__row_id'].astype(int).tolist())})
            closed,score=prototype_scores_from_z(pack,z_query,z_pool[ids],novel)
            for cal,cut in cuts.items():
                pred=closed.copy();pred[score>cut]='Unknown'
                stats=score_predictions(truth,pred,score,known+[novel],closed)
                old_f1=f1_score(truth[old_mask],pred[old_mask],labels=known,
                                average='macro',zero_division=0)
                base={'source':src,'target':tgt,'seed':seed,'method':method,'calibration':cal,
                      'stage':'post','novel_family':novel,'shots':k,'episode':ep,
                      'threshold':cut,
                      'target_known_calibration_n':len(calibration) if cal=='target_known' else 0,
                      'target_unlabeled_n':len(t_support) if cal=='target_unlabeled_gmm' else 0}
                RESULTS.append({**base,'novel_recall':float(np.mean(pred[novel_mask]==novel)),
                    'remaining_unknown_recall':float(np.mean(pred[remaining]=='Unknown')),
                    'old_known_macro_f1':float(old_f1),
                    'old_known_f1_change':float(old_f1-pre_old_f1[cal]),**stats})
                if k==5: FAMILY.extend(family_rows(truth,pred,score,known+[novel],base))
                if ((src,tgt,seed)==anchor_key and method=='RobustProto' and
                    novel==pair.unknown_choices[0] and k==5 and ep==0 and cal=='target_known'):
                    save_predictions(OUT/f'{run_id}_example_post.parquet',t_query,pred,score,'post',novel)
                    save_confusion(OUT/'example_post_confusion.png',truth,pred,known+[novel],
                                   f'{src} to {tgt}: target-known, 5 shots of {novel}')
    COMPUTE.append({'source':src,'target':tgt,'seed':seed,'method':'all_methods',
      'measurement':'pair_seed_total','wall_hours':(time.perf_counter()-tic)/3600,
      'device':torch.cuda.get_device_name(0) if DEVICE.type=='cuda' else str(DEVICE),
      'energy_note':'Elapsed wall time is not electrical energy or GPU utilization'})
    payload={'config_hash':config_hash,'results':RESULTS[starts[0]:],
             'family':FAMILY[starts[1]:],'history':TRAIN_HISTORY[starts[2]:],
             'support':EPISODE_SUPPORT[starts[3]:],'compute':COMPUTE[starts[4]:],
             'oracle':ORACLE[starts[5]:],'calibration_audit':CALIBRATION_AUDIT[starts[6]:],
             'rule_budget':RULE_BUDGET[starts[7]:],'openmax_audit':OPENMAX_AUDIT[starts[8]:],
             'logreg_audit':LOGREG_AUDIT[starts[9]:],
             'representation':REPRESENTATION[starts[10]:],
             'predictors':PREDICTORS[starts[11]:]}
    checkpoint.write_text(json.dumps(payload,default=lambda v:v.item() if hasattr(v,'item') else str(v)))
    for name,buffer in [('experiment_results',RESULTS),('per_family_results',FAMILY),
                        ('training_history',TRAIN_HISTORY),('episode_support_ids',EPISODE_SUPPORT),
                        ('compute_cost',COMPUTE),('oracle_threshold_gap',ORACLE),
                        ('unlabeled_calibration_audit',CALIBRATION_AUDIT),
                        ('threshold_rule_budget_surface',RULE_BUDGET),('openmax_fit_audit',OPENMAX_AUDIT),
                        ('logreg_convergence_audit',LOGREG_AUDIT),
                        ('representation_transfer',REPRESENTATION),
                        ('source_side_predictors',PREDICTORS)]:
        pd.DataFrame(buffer).to_csv(OUT/f'{name}.csv',index=False)
    print('Finished in',round((time.perf_counter()-tic)/60,2),'minutes')

results=pd.DataFrame(RESULTS)
post_methods=('ProtoCE','RobustProto','EpisodicProto','CORALProto')
required_pre_methods={'RandomForest','ProtoCE','EpisodicProto','MSP','Energy',
                      'OpenMaxVariant','RobustProto','CORALProto'}
calibrations={'source','target_unlabeled_gmm','target_known'}
representation_frame=pd.DataFrame(REPRESENTATION)
logreg_frame=pd.DataFrame(LOGREG_AUDIT)
actual_pre=int(results.stage.eq('pre').sum());actual_post=int(results.stage.eq('post').sum())
assert len(results)==actual_pre+actual_post and {'pre','post'}==set(results.stage)
expected_pre=expected_post=0
post_expected_by_method={method:0 for method in post_methods}
pair_seed_coverage=[]
for pair in eligible.itertuples(index=False):
  for seed in SEEDS:
    key=(pair.source,pair.target,seed)
    subset=results[(results.source==pair.source)&(results.target==pair.target)&
                   (results.seed==seed)]
    rep=representation_frame[(representation_frame.source==pair.source)&
                             (representation_frame.target==pair.target)&
                             (representation_frame.seed==seed)]
    audit=logreg_frame[(logreg_frame.source==pair.source)&
                       (logreg_frame.target==pair.target)&(logreg_frame.seed==seed)]
    assert len(audit)>0, ('LogReg fit audit missing',key)
    methods=required_pre_methods|({'LogReg'} if audit.converged.any() else set())
    assert len(rep)==len(methods) and set(rep.method)==methods, ('representation coverage',key)
    pre=subset[subset.stage.eq('pre')];post=subset[subset.stage.eq('post')]
    assert len(pre)==3*len(methods) and set(pre.method)==methods, ('pre coverage',key)
    assert pre.groupby('method').calibration.apply(
        lambda values:set(values)==calibrations).all(), ('pre calibrations',key)
    assert not pre.duplicated(['method','calibration']).any(), ('duplicate pre row',key)
    post_per_method=len(pair.unknown_choices)*len(CFG.shots)*EPISODES*len(calibrations)
    assert len(post)==post_per_method*len(post_methods) and set(post.method)==set(post_methods), ('post coverage',key)
    assert not post.duplicated(['method','novel_family','shots','episode','calibration']).any(), ('duplicate post row',key)
    for method in post_methods:
        method_post=post[post.method.eq(method)]
        assert len(method_post)==post_per_method, ('post method missing',key,method)
        assert set(method_post.novel_family)==set(pair.unknown_choices), ('post family missing',key,method)
        assert set(method_post.shots)==set(CFG.shots), ('post shot missing',key,method)
        assert set(method_post.episode)==set(range(EPISODES)), ('post episode missing',key,method)
        assert set(method_post.calibration)==calibrations, ('post calibration missing',key,method)
        post_expected_by_method[method]+=post_per_method
    expected_pre+=3*len(methods);expected_post+=post_per_method*len(post_methods)
    pair_seed_coverage.append({'source':pair.source,'target':pair.target,'seed':seed,
        'pre_methods':sorted(methods),'pre_rows':len(pre),'post_rows':len(post),
        'post_rows_per_method':post_per_method,'logreg_converged':bool(audit.converged.any())})
assert (actual_pre,actual_post)==(expected_pre,expected_post),(
    actual_pre,expected_pre,actual_post,expected_post)
post_actual_by_method={m:int((results.stage.eq('post')&results.method.eq(m)).sum())
                       for m in post_methods}
assert post_actual_by_method==post_expected_by_method
(OUT/'experiment_row_coverage.json').write_text(json.dumps({
    'actual_pre':actual_pre,'expected_pre':expected_pre,
    'actual_post':actual_post,'expected_post':expected_post,
    'post_rows_per_method_actual':post_actual_by_method,
    'post_rows_per_method_expected':post_expected_by_method,
    'source_target_pairs':len(eligible),'seeds':len(SEEDS),
    'pair_seed_coverage':pair_seed_coverage,
    'post_methods':list(post_methods),
    'post_rows_depend_on_novel_family_count_per_pair':True},indent=2))
for (src,tgt,seed),attempts in pd.DataFrame(LOGREG_AUDIT).groupby(['source','target','seed']):
    logged=results[(results.source==src)&(results.target==tgt)&
                   (results.seed==seed)&(results.method=='LogReg')]
    assert bool(attempts.converged.any())==bool(len(logged)), (src,tgt,seed)
    assert not ((attempts.converged)&
                (attempts.max_iterations_used>=attempts.max_iter)).any()
# Verify the actual surface: B=0 has one rule, and each feasible B>0 has three.
coverage_rows=[]
for key,rows in pd.DataFrame(RULE_BUDGET).groupby(['source','target','seed','method']):
    for budget in (0,5,10,20,50,'all'):
        part=rows[rows.budget_per_known_class.eq(budget)]
        expected={'source_only'} if budget==0 else {'pooled_q','benign_q','max_guard'}
        completed=part[part.status.eq('completed')]
        infeasible=part[part.status.eq('infeasible')]
        assert (len(completed)==len(expected) and set(completed['rule'])==expected
                and len(infeasible)==0) or (budget!=0 and len(completed)==0
                                            and len(infeasible)==1), (key,budget)
        coverage_rows.append(dict(zip(('source','target','seed','method'),key),
            budget_per_known_class=budget,status='completed' if len(completed) else 'infeasible',
            completed_rows=len(completed),expected_if_feasible=len(expected)))
pd.DataFrame(coverage_rows).to_csv(OUT/'rule_budget_coverage.csv',index=False)
assert first_example is not None
assert (first_example['source'],first_example['target'],first_example['seed'])==anchor_key
display(results[results.stage=='pre'][['source','target','method','calibration','known_macro_f1',
       'unknown_auroc','unknown_macro_auroc','benign_alert_rate']].head(25))
```

## Task 9. Ablation, conflict sensitivity, and host-shift validation
The conflict experiment compares quarantine with retaining all repeated/conflicting flows **as indivisible feature groups**. Its target query composition changes, so the contrast is a sensitivity analysis, not a paired treatment effect. A group-level permutation of source training labels on the anchor pair is a limited negative control: this verifies that the pipeline uses labels, but cannot establish temporal validity. A second experiment attempts full host-disjoint splitting by connected components of both IP endpoints. If infeasible, it independently attempts source-IP-disjoint splitting and reports any destination overlap. IPs are only split metadata, never model inputs. Endpoint connected components are built only on capped samples and use memory proportional to their distinct endpoint count; do not apply this Python implementation directly to raw multi-million-flow captures. The uploaded v2 files lack verified time columns; a chronological experiment requires timestamp-bearing data and is not imputed here.

```python
ex=first_example
ABLATIONS=[]
for seed in (SEEDS if CFG.mode=='FINAL' else SEEDS[:1]):
    known,s_train,s_dev,t_support,t_query,pre=prepare_pair(ex['source'],ex['target'],seed)
    train_x=pre.transform(s_train);train_y=s_train['__label'].map({v:i for i,v in enumerate(known)}).to_numpy(dtype=np.int64)
    novel=ex['novel'];pool=t_support[t_support['__label'].eq(novel)]
    x_query=pre.transform(t_query);truth=t_query.__label.to_numpy(dtype=object)
    mask=truth==novel
    for kind in ('NoContrast','NoAugment'):
        pack=fit_encoder(train_x,train_y,known,seed,kind)
        cut,_=target_threshold(pack,pre,t_support,known,CFG.target_calibration_per_known_class,
                               seed,CFG.known_acceptance_quantile)
        zq=embeddings(pack['model'],x_query);zs=embeddings(pack['model'],pre.transform(pool))
        for ep in range(EPISODES):
            ids=episode_indices(ex['source'],ex['target'],seed,novel,5,ep,len(pool))
            closed,score=prototype_scores_from_z(pack,zq,zs[ids],novel)
            pred=closed.copy();pred[score>cut]='Unknown'
            ABLATIONS.append({'source':ex['source'],'target':ex['target'],'seed':seed,
                'method':kind,'episode':ep,'novel_family':novel,'shots':5,
                'novel_recall':float(np.mean(pred[mask]==novel)),
                'known_macro_f1':float(f1_score(truth[np.isin(truth,known)],
                    pred[np.isin(truth,known)],labels=known,average='macro',zero_division=0))})
pd.DataFrame(ABLATIONS).to_csv(OUT/'ablation_episode_results.csv',index=False)

# Group-level source-label permutation control on the anchor pair. Permuting
# within exact-feature groups would do nothing for singleton deduplicated groups.
NEGATIVE_CONTROL=[]
anchor_true=results[(results.source==ex['source'])&(results.target==ex['target'])&
                    (results.seed==ex['seed'])&(results.method=='RobustProto')&
                    (results.stage=='pre')&(results.calibration=='source')].iloc[0]
NEGATIVE_CONTROL.append({'control':'true_source_labels','permutation':-1,
    'source':ex['source'],'target':ex['target'],'seed':ex['seed'],
    'known_macro_f1':float(anchor_true.known_macro_f1),
    'unknown_auroc':float(anchor_true.unknown_auroc),
    'benign_alert_rate':float(anchor_true.benign_alert_rate)})
group_labels=ex['s_train'].groupby('__group')['__label'].first()
for perm in range(3 if CFG.mode=='PILOT' else 10):
    shuffled=np.random.default_rng(10000+perm).permutation(group_labels.to_numpy())
    shuffled_map=dict(zip(group_labels.index,shuffled))
    control_y=ex['s_train']['__group'].map(shuffled_map).map(
        {name:i for i,name in enumerate(ex['known'])}).to_numpy(dtype=np.int64)
    assert np.array_equal(np.bincount(control_y),np.bincount(ex['y_train']))
    null_pack=fit_encoder(ex['x_train'],control_y,ex['known'],ex['seed']+10000+perm,'RobustProto')
    null_cut=calibrate_threshold(null_pack,ex['pre'].transform(ex['s_dev']),CFG.known_acceptance_quantile)
    stats,_,_=evaluate(null_pack,null_cut,ex['pre'],ex['t_query'],ex['known'])
    NEGATIVE_CONTROL.append({'control':'group_label_permutation','permutation':perm,
        'source':ex['source'],'target':ex['target'],'seed':ex['seed'],
        'known_macro_f1':stats['known_macro_f1'],
        'unknown_auroc':stats['unknown_auroc'],'benign_alert_rate':stats['benign_alert_rate']})
pd.DataFrame(NEGATIVE_CONTROL).to_csv(OUT/'group_label_negative_control.csv',index=False)

SENSITIVITY=[]
retained_pairs=pair_registry(RETAINED)
nb_pairs=pd.concat([PAIRS[PAIRS.eligible & PAIRS.source.eq('NB15')].head(1),
                    PAIRS[PAIRS.eligible & PAIRS.target.eq('NB15')].head(1)],ignore_index=True)
for pair in nb_pairs.itertuples(index=False):
  for seed in (SEEDS if CFG.mode=='FINAL' else SEEDS[:1]):
    for policy,samples,pairs in [('quarantine',SAMPLES,PAIRS),('retain_grouped',RETAINED,retained_pairs)]:
      try:
        known,s_train,s_dev,t_support,t_query,pre=prepare_pair(pair.source,pair.target,seed,
                   samples=samples,pairs=pairs,write_manifest=False)
        x=pre.transform(s_train);y=s_train.__label.map({v:i for i,v in enumerate(known)}).to_numpy(dtype=np.int64)
        pack=fit_encoder(x,y,known,seed,'RobustProto')
        cut,_=target_threshold(pack,pre,t_support,known,CFG.target_calibration_per_known_class,
                               seed,CFG.known_acceptance_quantile)
        metrics,_,_=evaluate(pack,cut,pre,t_query,known)
        SENSITIVITY.append({'source':pair.source,'target':pair.target,'policy':policy,
            'seed':seed,'source_train_n':len(s_train),'target_query_n':len(t_query),
            'status':'completed',**metrics})
      except ValueError as err:
        SENSITIVITY.append({'source':pair.source,'target':pair.target,'policy':policy,
                             'seed':seed,'status':'infeasible','reason':str(err)})
pd.DataFrame(SENSITIVITY).to_csv(OUT/'conflict_policy_sensitivity.csv',index=False)

def host_component_diagnostic(frame,role,src,tgt):
    components=endpoint_components(frame)
    counts=pd.Series(components).value_counts()
    x=frame[['__label']].copy();x['component']=components
    by_class=x.groupby('__label')['component'].nunique()
    return {'source':src,'target':tgt,'role':role,'n_flows':len(frame),
            'n_endpoint_components':len(counts),
            'largest_component_flows':int(counts.max()),
            'largest_component_fraction':float(counts.max()/len(frame)),
            'components_per_class':json.dumps(by_class.astype(int).to_dict()),
            'strict_split_structurally_possible':bool((by_class>=2).all()),
            'note':'per-class coverage is necessary but not sufficient for the fixed support/query budgets'}

HOST_VALIDATION=[];HOST_COMPONENTS=[]
host_candidates=eligible if CFG.mode=='FINAL' else eligible.head(4)
for pair in host_candidates.itertuples(index=False):
  src,tgt=pair.source,pair.target
  try:
      source_frame=SAMPLES[src][SAMPLES[src].__label.isin(pair.known)].copy()
      target_frame=SAMPLES[tgt][SAMPLES[tgt].__label.isin(
          list(pair.known)+list(pair.unknown_choices))].copy()
      target_frame=target_frame[~target_frame.__group.isin(set(source_frame.__group))]
      HOST_COMPONENTS.extend((host_component_diagnostic(source_frame,'source',src,tgt),
                              host_component_diagnostic(target_frame,'target',src,tgt)))
  except (ValueError,KeyError) as err:
      HOST_COMPONENTS.append({'source':src,'target':tgt,'role':'both',
                              'strict_split_structurally_possible':False,'reason':str(err)})
  for split_mode in ('host','source_host'):
    try:
      known,s_train,s_dev,t_support,t_query,pre=prepare_pair(pair.source,pair.target,SEEDS[0],
                                                   split_mode=split_mode)
      x=pre.transform(s_train);y=s_train.__label.map({v:i for i,v in enumerate(known)}).to_numpy(dtype=np.int64)
      pack=fit_encoder(x,y,known,SEEDS[0],'RobustProto')
      cut,_=target_threshold(pack,pre,t_support,known,CFG.target_calibration_per_known_class,
                             SEEDS[0],CFG.known_acceptance_quantile)
      metrics,_,_=evaluate(pack,cut,pre,t_query,known)
      endpoints=lambda f:set(f.IPV4_SRC_ADDR.astype(str))|set(f.IPV4_DST_ADDR.astype(str))
      overlap=len(endpoints(t_support)&endpoints(t_query))
      HOST_VALIDATION.append({'source':pair.source,'target':pair.target,'seed':SEEDS[0],
        'split_mode':split_mode,'status':'completed','source_IP_overlap':0,
        'all_endpoint_overlap':overlap,'fully_host_disjoint':overlap==0,**metrics})
    except (ValueError,KeyError) as err:
      HOST_VALIDATION.append({'source':pair.source,'target':pair.target,
        'seed':SEEDS[0],'split_mode':split_mode,'status':'infeasible','reason':str(err)})
pd.DataFrame(HOST_VALIDATION).to_csv(OUT/'host_split_validation.csv',index=False)
pd.DataFrame(HOST_COMPONENTS).to_csv(OUT/'host_components_audit.csv',index=False)
display(pd.DataFrame(SENSITIVITY));display(pd.DataFrame(HOST_VALIDATION))
```

## Task 10. Hierarchical summaries and reviewer-facing outputs
Episode intervals quantify uncertainty in *support selection conditional on that fixed target query*, not independent dataset evidence. Pair-level summaries average episodes first, then families, then seeds. Directed pairs reuse the same four underlying captures, so bootstrap intervals across them are descriptive and should never be called independent-site inference.
```python
pre_results=results[results.stage=='pre'].copy()
post_results=results[results.stage=='post'].copy()
for name,frame in (('pre',pre_results),('post',post_results)):
    assert np.allclose(frame.benign_alert_rate,
        frame.benign_rejection_rate+frame.benign_accepted_misclassification_rate,
        atol=1e-10), (name,'observed benign-alert decomposition')
    assert np.allclose(frame.benign_alert_rate,
        frame.benign_closed_error_floor+frame.benign_correct_closed_rejection_rate,
        atol=1e-10), (name,'closed-set benign-alert decomposition')
pair_pre=pre_results.groupby(['source','target','method','calibration'],as_index=False)[
  ['known_macro_f1','unknown_recall','unknown_macro_recall','unknown_auroc',
   'unknown_macro_auroc','unknown_aupr','oscr','benign_alert_rate',
   'benign_rejection_rate','benign_accepted_misclassification_rate',
   'benign_closed_error_floor','benign_correct_closed_rejection_rate']].mean()
assert np.allclose(pair_pre['benign_alert_rate'],
    pair_pre['benign_rejection_rate']+
    pair_pre['benign_accepted_misclassification_rate'],atol=1e-10)
assert np.allclose(pair_pre['benign_alert_rate'],
    pair_pre['benign_closed_error_floor']+
    pair_pre['benign_correct_closed_rejection_rate'],atol=1e-10)
pair_pre.to_csv(OUT/'pre_pair_summary.csv',index=False)
pd.DataFrame(REPRESENTATION).to_csv(OUT/'representation_transfer.csv',index=False)

# Prequery indicators are available without target labels. The outcome below
# requires target query labels and is only associated post hoc. Four reused
# captures and candidate selection preclude a deployment switch claim.
predictor_frame=pd.DataFrame(PREDICTORS)
for name in ('pseudo_class_prototype_cosine_shift','geometry_unavailable',
             'geometry_unavailable_reason','n_pseudo_classes_with_centroids'):
    assert predictor_frame.groupby(['source','target','seed'])[name].nunique(
        dropna=False).eq(1).all(), ('shared geometry differs across methods',name)
assert (predictor_frame.geometry_unavailable==
        predictor_frame.pseudo_class_prototype_cosine_shift.isna()).all()
geometry_coverage=predictor_frame[['source','target','seed','geometry_unavailable',
    'geometry_unavailable_reason','n_pseudo_classes_with_centroids']].drop_duplicates()
assert not geometry_coverage.duplicated(['source','target','seed']).any()
geometry_coverage['geometry_available']=~geometry_coverage.geometry_unavailable
geometry_coverage.to_csv(OUT/'geometry_coverage.csv',index=False)
predictor_columns=['score_wasserstein_source_iqr_units',
    'predicted_benign_score_wasserstein_source_iqr_units',
    'pseudo_class_prototype_cosine_shift','source_tail_exceedance_delta']
from scipy.stats import spearmanr
predictor_pairs=predictor_frame.groupby(['source','target','method'],as_index=False)[
    predictor_columns+['source_threshold_oracle_gap_diagnostic_only']].mean()
geometry_seed_counts=predictor_frame.groupby(['source','target','method'],as_index=False)[
    'pseudo_class_prototype_cosine_shift'].count().rename(columns={
        'pseudo_class_prototype_cosine_shift':'n_geometry_seeds_available'})
predictor_pairs=predictor_pairs.merge(geometry_seed_counts,on=['source','target','method'],
                                      validate='one_to_one')
assert (predictor_pairs.pseudo_class_prototype_cosine_shift.notna()==
        predictor_pairs.n_geometry_seeds_available.gt(0)).all()
predictor_pairs.to_csv(OUT/'source_side_predictors_by_pair.csv',index=False)
predictor_correlations=[]
for method,part in predictor_pairs.groupby('method'):
    for name in predictor_columns:
        valid=part[[name,'source_threshold_oracle_gap_diagnostic_only']].dropna()
        rho=(float(spearmanr(valid[name],
             valid.source_threshold_oracle_gap_diagnostic_only).statistic)
             if len(valid)>=3 and valid[name].nunique()>1 and
             valid.source_threshold_oracle_gap_diagnostic_only.nunique()>1 else np.nan)
        predictor_correlations.append({'method':method,'indicator':name,'n_directed_pairs':len(valid),
            'spearman_rho_descriptive_only':rho,
            'status':'exploratory post-hoc association; query labels used for outcome; reused captures'})
correlation_frame=pd.DataFrame(predictor_correlations)
correlation_frame.to_csv(
    OUT/'predictor_exploratory_correlations.csv',index=False)
view=predictor_pairs[predictor_pairs.method.eq('RobustProto')]
fig,axes=plt.subplots(2,2,figsize=(11,8))
for axis,name in zip(axes.flat,predictor_columns):
    for row in view.itertuples(index=False):
        x=getattr(row,name);y=row.source_threshold_oracle_gap_diagnostic_only
        if np.isfinite(x) and np.isfinite(y):
            axis.scatter(x,y,color='tab:blue')
            axis.annotate(f'{row.source}→{row.target}',(x,y),fontsize=8)
    axis.set(xlabel=name.replace('_',' '),ylabel='Source oracle gap (query-label diagnostic)')
fig.suptitle('Exploratory transfer indicators; directed pairs share four captures')
fig.tight_layout();fig.savefig(OUT/'source_side_predictors_exploratory.png',dpi=180);plt.close(fig)

# Headline threshold-transfer table: matched source/target query and model. The
# oracle utility is a diagnostic ceiling selected with query labels after evaluation.
gap=pd.DataFrame(ORACLE).groupby(['source','target','method','calibration'],as_index=False)[
    ['threshold_utility','oracle_utility','oracle_gap']].mean()
transfer=pair_pre.merge(gap,on=['source','target','method','calibration'],validate='one_to_one')
transfer.to_csv(OUT/'threshold_transfer_by_pair.csv',index=False)
seed_utility=pd.DataFrame(ORACLE)
seed_utility=seed_utility[seed_utility.stage.eq('pre')].pivot(
    index=['source','target','seed','method'],columns='calibration',
    values='threshold_utility').rename(columns={'source':'source_cut_utility'}).reset_index()
def seed_conditional_interval(src,tgt,method,adaptation):
    rows=seed_utility[(seed_utility.source==src)&(seed_utility.target==tgt)&
                      (seed_utility.method==method)].sort_values('seed')
    deltas=(rows[adaptation]-rows['source_cut_utility']).dropna().to_numpy(dtype=float)
    if len(deltas)<2: return len(deltas),np.nan,np.nan
    key=f'v8-seed-bootstrap|{src}|{tgt}|{method}|{adaptation}'.encode()
    rng=np.random.default_rng(int.from_bytes(hashlib.sha256(key).digest()[:8],'little'))
    means=rng.choice(deltas,size=(CFG.episode_ci_bootstrap_reps,len(deltas)),
                     replace=True).mean(1)
    lo,hi=np.quantile(means,[.025,.975])
    return len(deltas),float(lo),float(hi)
delta_rows=[]
for (src,tgt,method),piece in transfer.groupby(['source','target','method']):
    lookup=piece.set_index('calibration')
    if not {'source','target_unlabeled_gmm','target_known'}.issubset(lookup.index): continue
    for cal in ('target_unlabeled_gmm','target_known'):
        n_seeds,ci_lo,ci_hi=seed_conditional_interval(src,tgt,method,cal)
        delta_rows.append({'source':src,'target':tgt,'method':method,'adaptation':cal,
            'delta_utility_vs_source':float(lookup.loc[cal,'threshold_utility']-lookup.loc['source','threshold_utility']),
            'delta_benign_alert_vs_source':float(lookup.loc[cal,'benign_alert_rate']-lookup.loc['source','benign_alert_rate']),
            'delta_unknown_recall_vs_source':float(lookup.loc[cal,'unknown_recall']-lookup.loc['source','unknown_recall']),
            'oracle_gap_after_adaptation':float(lookup.loc[cal,'oracle_gap']),
            'n_training_seeds':n_seeds,'seed_conditional_delta_utility_ci_lo':ci_lo,
            'seed_conditional_delta_utility_ci_hi':ci_hi,
            'ci_scope':'seed resampling conditional on the same pair/query; pilot n=1 has no interval'})
pd.DataFrame(delta_rows).to_csv(OUT/'threshold_transfer_matched_deltas.csv',index=False)
# Step 1: reopen the three saved files. Their schema and the numerical
# reviewer note now depend on the exact CSVs the user will inspect in Drive.
rep_frame=pd.read_csv(OUT/'representation_transfer.csv')
correlation_frame=pd.read_csv(OUT/'predictor_exploratory_correlations.csv')
delta_frame=pd.read_csv(OUT/'threshold_transfer_matched_deltas.csv')
assert {'source','target','method','source_dev_closed_macro_f1',
        'target_query_known_closed_macro_f1'}.issubset(rep_frame.columns)
assert {'method','indicator','n_directed_pairs',
        'spearman_rho_descriptive_only'}.issubset(correlation_frame.columns)
assert {'source','target','method','adaptation',
        'delta_utility_vs_source'}.issubset(delta_frame.columns)

# The evidence note answers the reviewer using the selected RobustProto
# comparison, and also discloses how the answer changes across methods.
# These sentences are generated only after the actual run has its tables.
robust_rep=rep_frame[rep_frame.method.eq('RobustProto')].groupby(
    ['source','target'],as_index=False)[
    ['source_dev_closed_macro_f1','target_query_known_closed_macro_f1']].mean()
robust_rep['closed_set_f1_loss']=(
    robust_rep.source_dev_closed_macro_f1-robust_rep.target_query_known_closed_macro_f1)
rep_lookup={(r.source,r.target):r for r in robust_rep.itertuples(index=False)}
bot_gap=rep_lookup.get(('BOT','NB15'));ton_gap=rep_lookup.get(('TON','CIC'))
if bot_gap is not None and ton_gap is not None:
    gap_sentence=(f'RobustProto source-to-target closed-set F1 losses are '
        f'{ton_gap.closed_set_f1_loss:.4f} for TON→CIC and '
        f'{bot_gap.closed_set_f1_loss:.4f} for BOT→NB15 '
        f'(TON→CIC minus BOT→NB15 = '
        f'{ton_gap.closed_set_f1_loss-bot_gap.closed_set_f1_loss:+.4f}). '
        f'TON→CIC is {"larger" if ton_gap.closed_set_f1_loss>bot_gap.closed_set_f1_loss else "not larger"} '
        'in this comparison. No numeric threshold for “much larger” was prespecified. '
        'These are total closed-set transfer differences; they do not isolate the encoder.')
else: gap_sentence='One of the two requested RobustProto directed pairs was not completed.'
selected_pairs=[(str(r.source),str(r.target)) for r in eligible.itertuples(index=False)]
selected_sentence='Executed directed pairs: '+', '.join(f'{src}→{tgt}' for src,tgt in selected_pairs)+'.'
if bot_gap is None or ton_gap is None:
    missing=[f'{src}→{tgt}' for src,tgt in (('BOT','NB15'),('TON','CIC'))
             if (src,tgt) not in rep_lookup]
    gap_sentence=('The requested RobustProto comparison is not estimable from this run: '
                  +', '.join(missing)+' absent. The actual selected pairs are listed above; '
                  'another pair is not a substitute for the requested contrast.')
gap_table=['| Pair | Source closed F1 | Target known closed F1 | Source minus target |',
           '|---|---:|---:|---:|']
gap_table += [f'| {r.source}→{r.target} | {r.source_dev_closed_macro_f1:.4f} | '
              f'{r.target_query_known_closed_macro_f1:.4f} | {r.closed_set_f1_loss:+.4f} |'
              for r in robust_rep.itertuples(index=False)]
candidate_rho=correlation_frame[(correlation_frame.n_directed_pairs>=4)&
                                 (correlation_frame.spearman_rho_descriptive_only>=.5)]
if len(candidate_rho):
    rows=candidate_rho.sort_values('spearman_rho_descriptive_only',ascending=False)
    indicator_sentence='Descriptive ρ≥0.5 occurs for '+', '.join(
        f"{r.method}/{r.indicator}: ρ={r.spearman_rho_descriptive_only:.3f} (n={r.n_directed_pairs})"
        for r in rows.itertuples(index=False))+'. No selector is validated.'
else:
    indicator_sentence='No method/indicator has descriptive ρ≥0.5 with at least four pairs.'
known_delta=delta_frame[delta_frame.adaptation.eq('target_known')]
robust_delta=known_delta[known_delta.method.eq('RobustProto')].sort_values(['source','target'])
positive=int((robust_delta.delta_utility_vs_source>0).sum())
delta_sentence=(f'RobustProto target-known versus source utility improves on '
    f'{positive}/{len(robust_delta)} executed directed pairs'
    f'{" (four-pair question answered)" if len(robust_delta)==4 else " (four-pair question not estimable)"}: '+', '.join(
       f'{r.source}→{r.target} {r.delta_utility_vs_source:+.4f}'
       for r in robust_delta.itertuples(index=False))+'.')
bot_baseline=pair_pre[(pair_pre.source.eq('BOT'))&(pair_pre.target.eq('NB15'))&
                      (pair_pre.calibration.eq('source'))&
                      (pair_pre.method.isin(['LogReg','RobustProto']))]
if set(bot_baseline.method)=={'LogReg','RobustProto'}:
    bot_metrics=bot_baseline.set_index('method')
    lr=bot_metrics.loc['LogReg'];rp=bot_metrics.loc['RobustProto']
    logreg_sentence=(f'BOT→NB15 source-threshold LogReg converged: known macro F1 '
        f'{lr.known_macro_f1:.4f} versus RobustProto {rp.known_macro_f1:.4f}; '
        f'unknown AUROC {lr.unknown_auroc:.4f} versus {rp.unknown_auroc:.4f}. '
        'State any negative finding separately by metric.')
else:
    logreg_sentence=('BOT→NB15 LogReg comparison is unavailable in this run '
        '(pair absent or LogReg did not converge). Do not cite a V8 LogReg finding for this pair.')
method_gain=(known_delta.groupby('method').delta_utility_vs_source.agg(
    n_pairs='size',n_positive=lambda s:int((s>0).sum())).reset_index())
geometry_available=int(geometry_coverage.geometry_available.sum())
geometry_total=len(geometry_coverage)
notes=[f'# {CFG.mode.title()} evidence notes', '',
    'Generated from the current run’s representation_transfer.csv, '
    'predictor_exploratory_correlations.csv and threshold_transfer_matched_deltas.csv.',
    '',selected_sentence,'',
    'Shared ProtoCE geometry available in '
    f'{geometry_available}/{geometry_total} directed pair × seed runs. '
    'See geometry_coverage.csv for missing-value reasons; exploratory correlations '
    'exclude missing pair/method indicators and report n_directed_pairs.','',
    '## 1. Closed-set transfer', '',gap_sentence,'',*gap_table,'',
    '## 2. Prequery indicator association with a query-label diagnostic','',
    indicator_sentence,
    'Multiple candidates and four reused captures make these exploratory associations. '
    'They do not select a source/GMM deployment rule.','',
    '## 3. Target-known utility gain relative to source','',
    delta_sentence,'',
    'Counts by method (directed pairs; positive deltas only):','',
    '| Method | Positive / evaluated |','|---|---:|']
notes += [f'| {r.method} | {r.n_positive} / {r.n_pairs} |'
          for r in method_gain.itertuples(index=False)]
notes += ['', '## 4. Baseline claim coverage','',logreg_sentence,'',
    '## FINAL scope decision','',
    f'Configured option A: {int(PAIRS.eligible.sum())} eligible directed pairs, '
    f'{len(CFG.seeds_final)} seeds and {CFG.episodes_final} support episodes per condition. '
    'This is an intended protocol, not an executed FINAL result. GPU quota has not '
    'been verified; measure runtime from a representative FINAL-scale job before '
    'reserving time. The reviewer’s 15–20 hour estimate is unverified.','',
    'Freeze FINAL scope only after reviewing these actual PILOT findings, the '
    'collision/capping sensitivity and the independent-validation limitations.']
(OUT/'notes.md').write_text('\n'.join(notes)+'\n')

# Method-specific manuscript draft: the alert identity is exact, while the
# score ranking and labeled target benefit can change independently of it.
paragraphs=['# Per-pair results narrative — descriptive draft','',
    'RobustProto, B=20 labeled target-known flows per known family. '
    'Scores and utilities use the held-out target query only for evaluation; '
    'this draft must be reviewed before publication.','']
robust_transfer=transfer[transfer.method.eq('RobustProto')]
for (src,tgt),part in robust_transfer.groupby(['source','target'],sort=True):
    at=part.set_index('calibration')
    if not {'source','target_unlabeled_gmm','target_known'}.issubset(at.index): continue
    s=at.loc['source'];g=at.loc['target_unlabeled_gmm'];k=at.loc['target_known']
    rejection_share=(f'{s.benign_rejection_rate/s.benign_alert_rate:.1%} of benign alerts'
                     if s.benign_alert_rate>0 else 'share undefined: zero benign alerts')
    paragraphs.append(
      f'**{src}→{tgt}.** Under the source threshold, the benign-alert rate was '
      f'{s.benign_alert_rate:.3f}: rejection {s.benign_rejection_rate:.3f} '
      f'({rejection_share}) and accepted known-class misclassification '
      f'{s.benign_accepted_misclassification_rate:.3f}. At B=20, benign rejection '
      f'changed by {k.benign_rejection_rate-s.benign_rejection_rate:+.3f}; '
      f'benign alerts were {k.benign_alert_rate:.3f}, including residual accepted '
      f'misclassification {k.benign_accepted_misclassification_rate:.3f}. '
      f'The fixed closed-set classifier has a benign-alert floor of '
      f'{k.benign_closed_error_floor:.3f}; rejection of benign rows that it '
      f'would have classified correctly contributes '
      f'{k.benign_correct_closed_rejection_rate:.3f}. '
      f'Unknown-family recall changed from {s.unknown_recall:.3f} to '
      f'{k.unknown_recall:.3f}. Unlabeled GMM yielded benign alerts '
      f'{g.benign_alert_rate:.3f}, unknown recall {g.unknown_recall:.3f}, and '
      f'threshold-utility change {g.threshold_utility-s.threshold_utility:+.3f} '
      f'relative to source. B=20 utility changed by '
      f'{k.threshold_utility-s.threshold_utility:+.3f}. '
      f'Threshold-independent unknown-family macro AUROC was '
      f'{s.unknown_macro_auroc:.3f}; '
      f'{"above" if s.unknown_macro_auroc>.5 else "at or below"} 0.5.\n')
(OUT/'paper_results_narrative.md').write_text('\n'.join(paragraphs)+'\n')

# Figure displays the exact alert decomposition for a named method, with all
# three calibrations in each pair rather than collapsing them into a grand mean.
robust_bars=pair_pre[pair_pre.method.eq('RobustProto')].sort_values(
    ['source','target','calibration']).copy()
labels=[f'{r.source}→{r.target} | {r.calibration}' for r in robust_bars.itertuples(index=False)]
fig,axis=plt.subplots(figsize=(10,max(5,len(labels)*.32)))
ypos=np.arange(len(labels))
axis.barh(ypos,robust_bars.benign_rejection_rate,label='Rejected as Unknown',color='#0072B2')
axis.barh(ypos,robust_bars.benign_accepted_misclassification_rate,
          left=robust_bars.benign_rejection_rate,
          label='Accepted as known attack',color='#D55E00')
axis.scatter(robust_bars.benign_closed_error_floor,ypos,marker='|',s=90,
             color='black',label='Fixed closed-set error floor',zorder=3)
axis.set(yticks=ypos,yticklabels=labels,xlabel='Fraction of benign query flows',
         xlim=(0,1),title='RobustProto benign-alert decomposition')
axis.invert_yaxis();axis.legend(loc='lower right')
fig.tight_layout();fig.savefig(OUT/'benign_alert_decomposition.png',dpi=200);plt.close(fig)

rule_frame=pd.DataFrame(RULE_BUDGET)
valid_rules=rule_frame[rule_frame.status.eq('completed')].copy()
valid_rules['budget_per_known_class']=valid_rules['budget_per_known_class'].astype(str)
rule_by_pair=valid_rules.groupby(['source','target','method','budget_per_known_class','rule'],
    as_index=False)[['benign_alert_rate','unknown_recall','known_macro_f1','threshold_utility','oracle_gap']].mean()
rule_by_pair.to_csv(OUT/'threshold_rule_budget_by_pair.csv',index=False)
if not rule_by_pair.empty:
    candidate=rule_by_pair[rule_by_pair.method.eq('RobustProto')].copy()
    candidate['budget_label']=candidate['budget_per_known_class'].astype(str)
    order=['0','5','10','20','50','all']
    trend=candidate.groupby(['budget_label','rule'],as_index=False)[
        ['benign_alert_rate','unknown_recall','known_macro_f1']].mean()
    fig,axes=plt.subplots(1,2,figsize=(13,5))
    for rule,piece in trend.groupby('rule'):
        ordered=piece.set_index('budget_label').reindex([b for b in order if b in set(piece.budget_label)])
        axes[0].plot(ordered.benign_alert_rate,ordered.unknown_recall,marker='o',label=rule)
        for budget,r in ordered.iterrows():
            axes[0].annotate(str(budget),(r.benign_alert_rate,r.unknown_recall),fontsize=8)
        axes[1].plot(ordered.index,ordered.known_macro_f1,marker='o',label=rule)
    axes[0].set(xlabel='Benign alert rate',ylabel='Unknown recall',title='Threshold-rule tradeoff (directed-pair means)')
    axes[1].set(xlabel='Known labels per family (all = entire support pool)',ylabel='Known macro F1',
                title='Known-family retention under target calibration')
    for axis in axes: axis.legend(fontsize=8)
    fig.tight_layout();fig.savefig(OUT/'threshold_rule_budget_tradeoff.png',dpi=180);plt.close(fig)
    fig,axis=plt.subplots(figsize=(7,5))
    for rule,piece in trend.groupby('rule'):
        ordered=piece.set_index('budget_label').reindex([b for b in order if b in set(piece.budget_label)])
        axis.plot(ordered.index,ordered.known_macro_f1,marker='o',label=rule)
    axis.set(xlabel='Labeled target-known flows per class (all = full support)',
             ylabel='Known macro F1',title='Known-class retention versus calibration budget')
    axis.legend();fig.tight_layout()
    fig.savefig(OUT/'known_macro_f1_by_budget.png',dpi=180);plt.close(fig)

episode_keys=['source','target','seed','method','calibration','novel_family','shots']
episode_rows=[];boot_rng=np.random.default_rng(20260928)
for key,part in post_results.groupby(episode_keys,sort=True):
    row=dict(zip(episode_keys,key));row['n_episodes']=len(part)
    for metric in ('novel_recall','remaining_unknown_recall','old_known_f1_change',
                   'known_macro_f1','benign_alert_rate'):
        vals=part[metric].dropna().to_numpy(dtype=float)
        means=boot_rng.choice(vals,size=(CFG.episode_ci_bootstrap_reps,len(vals)),replace=True).mean(axis=1)
        row[metric+'_mean']=float(vals.mean());row[metric+'_sd']=float(vals.std(ddof=1)) if len(vals)>1 else np.nan
        row[metric+'_episode_ci_lo'],row[metric+'_episode_ci_hi']=np.quantile(means,[.025,.975])
    episode_rows.append(row)
episode_summary=pd.DataFrame(episode_rows)
episode_summary.to_csv(OUT/'episode_mean_sd_ci.csv',index=False)
pair_post=episode_summary.groupby(['source','target','method','calibration','shots'],as_index=False)[
    ['novel_recall_mean','remaining_unknown_recall_mean','old_known_f1_change_mean',
     'known_macro_f1_mean','benign_alert_rate_mean']].mean()
pair_post.to_csv(OUT/'post_pair_summary.csv',index=False)

# Independent support draws at each shot count; the fixed query is reused.
# Bootstrap uncertainty concerns support draws only, not independent domains.
shot_rows=[];shot_rng=np.random.default_rng(271828)
group_cols=['source','target','seed','method','calibration','novel_family']
for keys,part in post_results.groupby(group_cols,sort=True):
    lo=part[part.shots.eq(1)].novel_recall.dropna().to_numpy(dtype=float)
    hi=part[part.shots.eq(10)].novel_recall.dropna().to_numpy(dtype=float)
    if not len(lo) or not len(hi): continue
    reps=CFG.episode_ci_bootstrap_reps
    changes=(shot_rng.choice(hi,size=(reps,len(hi)),replace=True).mean(1)-
             shot_rng.choice(lo,size=(reps,len(lo)),replace=True).mean(1))
    ci=np.quantile(changes,[.025,.975])
    shot_rows.append({**dict(zip(group_cols,keys)),'n_one_shot':len(lo),
        'n_ten_shot':len(hi),'ten_minus_one_mean_novel_recall':float(hi.mean()-lo.mean()),
        'support_bootstrap_ci_lo':float(ci[0]),'support_bootstrap_ci_hi':float(ci[1]),
        'scope':'independent support draws; fixed query; descriptive conditional interval'})
pd.DataFrame(shot_rows).to_csv(OUT/'shot_scaling_analysis.csv',index=False)
shot_view=post_results[(post_results.method.isin(['RobustProto','ProtoCE']))&
                       (post_results.calibration=='target_known')]
fig,axes=plt.subplots(1,2,figsize=(12,5),sharey=True)
for ax,(method,part) in zip(axes,shot_view.groupby('method')):
    jitter=np.random.default_rng(42).normal(0,.07,len(part))
    ax.scatter(part.shots.to_numpy()+jitter,part.novel_recall,alpha=.14,s=8)
    for shots,group in part.groupby('shots'):
        ax.scatter(shots,group.novel_recall.mean(),marker='D',s=55,color='black')
    ax.set(xlabel='Novel-family labeled shots',title=method,xticks=CFG.shots,ylim=(0,1))
axes[0].set_ylabel('Novel recall (episode; reused query)')
fig.tight_layout();fig.savefig(OUT/'shot_scaling_episodes.png',dpi=180);plt.close(fig)

def paired_pair_contrast(frame,metric,a='RobustProto',b='EpisodicProto',reps=1000):
    base=frame[frame.method.isin((a,b))]
    pivot=base.pivot(index=['source','target'],columns='method',values=metric).dropna()
    if pivot.empty or not {a,b}.issubset(pivot.columns):
        return {'metric':metric,'baseline':b,'n_directed_pairs':0}
    delta=(pivot[a]-pivot[b]).to_numpy();rng=np.random.default_rng(416)
    samples=np.mean(rng.choice(delta,size=(reps,len(delta)),replace=True),axis=1) if len(delta)>=3 else []
    return {'metric':metric,'baseline':b,'n_directed_pairs':len(delta),
            'mean_pair_delta':float(delta.mean()),
            'descriptive_ci_lo':float(np.quantile(samples,.025)) if len(samples) else np.nan,
            'descriptive_ci_hi':float(np.quantile(samples,.975)) if len(samples) else np.nan,
            'inference_note':'shared four datasets; intervals descriptive; no claim of independent sites'}

CONTRAST=[]
for cal in ('source','target_unlabeled_gmm','target_known'):
    f=pair_pre[pair_pre.calibration==cal]
    for b in ('ProtoCE','EpisodicProto','CORALProto','OpenMaxVariant','Energy','MSP','RandomForest','LogReg'):
        for metric in ('known_macro_f1','unknown_auroc','benign_alert_rate'):
            CONTRAST.append({'calibration':cal,**paired_pair_contrast(f,metric,b=b)})
pd.DataFrame(CONTRAST).to_csv(OUT/'paired_pair_contrasts.csv',index=False)

# Exploratory signed-rank tests retain the directed pair as the unit; treating
# hundreds of episodes as independent samples would grossly inflate power.
# Four shared capture origins also make nominal p-values descriptive only.
from scipy.stats import wilcoxon, norm
PAIR_TESTS=[]
for cal in ('source','target_unlabeled_gmm','target_known'):
    base=pair_pre[pair_pre.calibration==cal]
    for other in ('ProtoCE','EpisodicProto','CORALProto','OpenMaxVariant','Energy','MSP','RandomForest','LogReg'):
        for metric in ('known_macro_f1','unknown_auroc','benign_alert_rate'):
            piv=base[base.method.isin(('RobustProto',other))].pivot(
                index=['source','target'],columns='method',values=metric).dropna()
            if len(piv)<3 or not {'RobustProto',other}.issubset(piv.columns): continue
            differences=(piv['RobustProto']-piv[other]).to_numpy(dtype=float)
            nonzero=differences[differences!=0]
            p=float(wilcoxon(nonzero).pvalue) if len(nonzero)>=3 else np.nan
            # Rank-biserial is paired; Cliff's delta assumes independent samples.
            if len(nonzero):
                from scipy.stats import rankdata
                ranks=rankdata(abs(nonzero));effect=float(np.sum(np.sign(nonzero)*ranks)/sum(ranks))
            else: effect=0.
            PAIR_TESTS.append({'calibration':cal,'baseline':other,'metric':metric,
                'n_pairs':len(piv),'n_nonzero':len(nonzero),'median_pair_difference':float(np.median(differences)),
                'wilcoxon_p_unadjusted':p,'paired_rank_biserial':effect,
                'warning':'exploratory: directed pairs share capture origins; not independent-site inference'})
test_frame=pd.DataFrame(PAIR_TESTS)
if len(test_frame):
    test_frame['holm_p_within_calibration_metric']=np.nan
    for _,idx in test_frame.groupby(['calibration','metric']).groups.items():
        loc=np.asarray(list(idx));valid=loc[test_frame.loc[loc,'wilcoxon_p_unadjusted'].notna().to_numpy()]
        if len(valid):
            order=valid[np.argsort(test_frame.loc[valid,'wilcoxon_p_unadjusted'].to_numpy())]
            raw=test_frame.loc[order,'wilcoxon_p_unadjusted'].to_numpy()*np.arange(len(order),0,-1)
            test_frame.loc[order,'holm_p_within_calibration_metric']=np.minimum(1,np.maximum.accumulate(raw))
test_frame.to_csv(OUT/'exploratory_pair_tests.csv',index=False)

# Normal approximation to minimum detectable paired mean difference. It is a
# sensitivity calculation, not observed power or proof of adequate sample size.
power_rows=[]
for n in sorted(set(CONTRAST[i]['n_directed_pairs'] for i in range(len(CONTRAST)))):
    if n<2: continue
    for rho in (.3,.5,.8):
        power_rows.append({'n_directed_pairs':n,'assumed_pair_effect_sd':rho,
            'two_sided_alpha':.05,'target_power':.80,
            'approx_min_detectable_difference':float((norm.ppf(.975)+norm.ppf(.80))*rho/np.sqrt(n)),
            'note':'illustrative normal approximation; four domains reused; no independent-site power'})
pd.DataFrame(power_rows).to_csv(OUT/'pair_power_sensitivity.csv',index=False)

coverage=pd.DataFrame(HOST_VALIDATION).groupby('split_mode',as_index=False).agg(
    attempted=('status','size'),completed=('status',lambda s:int((s=='completed').sum())))
coverage['fraction_completed']=coverage['completed']/coverage['attempted']
coverage.to_csv(OUT/'host_validation_coverage.csv',index=False)

fig,axes=plt.subplots(1,2,figsize=(13,4.5))
for method in pair_pre.method.unique():
    p=pair_pre[(pair_pre.method==method)&(pair_pre.calibration=='target_known')]
    axes[0].scatter(p.unknown_auroc,p.benign_alert_rate,label=method,alpha=.75)
axes[0].set(xlabel='Unknown AUROC',ylabel='Benign alert rate',title='Individual directed pairs')
axes[0].legend(fontsize=7,ncol=2)
for method in ('ProtoCE','EpisodicProto','RobustProto','CORALProto'):
    p=pair_post[(pair_post.method==method)&(pair_post.calibration=='target_known')]
    curve=p.groupby('shots').novel_recall_mean.mean()
    axes[1].plot(curve.index,curve.values,marker='o',label=method)
axes[1].set(xlabel='Novel-family labeled shots',ylabel='Novel recall',ylim=(0,1),
            title='Mean across pair/family/seed episode means')
axes[1].legend(fontsize=8);fig.tight_layout();fig.savefig(OUT/'reviewer_summary.png',dpi=180);plt.close(fig)

readiness={
 'status':'PILOT — descriptive only' if CFG.mode=='PILOT' else 'FINAL run — methodological review still required',
 'known_target_calibration_labels_per_class':CFG.target_calibration_per_known_class,
 'target_threshold_rules':['pooled_q','benign_q','max_guard'],
 'budget_per_known_family':[0,5,10,20,50,'all_available'],
 'unlabeled_target_calibration':'two-component score GMM on disjoint but class-capped support; heuristic, not a deployment-prior guarantee',
 'oracle_diagnostic':'post hoc query labels used only in separate oracle_threshold_gap.csv; oracle threshold never evaluated as deployable',
 'novel_support_shots':list(CFG.shots),'episodes_per_condition':EPISODES,
 'directed_pairs':eligible[['source','target']].to_dict('records'),
 'novelty_statement':'SupCon + prototype addition is prior art. No new algorithm is claimed.',
 'temporal_validation':'unavailable: uploaded NetFlow v2 files have no verified timestamps; use documented timestamp-bearing captures to add it',
 'host_validation_status':pd.DataFrame(HOST_VALIDATION).status.value_counts().to_dict(),
 'v8_status':'pilot evidence notes and figures are generated from this run; FINAL remains unrun until preregistration',
 'geometry_indicator':'one shared source-trained ProtoCE value per pair/seed; missing geometry is recorded rather than terminating evaluation',
 'geometry_coverage_file':'geometry_coverage.csv',
 'bootstrap_seed_rule':'SHA-256 of source|target|method|adaptation, with sorted training seeds; conditional on one pair/query',
 'compute_note':'Repeated single-set and anchor-pair batched-set throughput are distinct; GPU energy is gross board energy when NVML supports it; emissions not inferred from time alone',
 'remaining_review':[
   'Inspect CIC mappings and class-by-class conflict losses; retain-grouped and quarantine target populations differ.',
   'Verify that calibration label budget and target-domain availability match the deployment threat model.',
   'Explain high benign alert and low unknown AUROC where observed; include negative findings.',
   'Add independent temporal/session data before describing attacks as chronologically emerging.',
   'Add a source-only CORAL/domain-adaptation baseline and an independent capture for a stronger threshold-transfer paper.',
   'Review licences and raw source/destination IP addresses for privacy; model feature exclusion does not erase raw addresses.',
   'Do not fit a GMM/source switching threshold from four query-labeled pilot pairs and call it deployed; validate on independent captures.',
   'Cite originals and NetFlow release; preregister FINAL hypotheses and stopping rule; document measured compute.',
   'Only run FINAL after protocol is frozen; Q1 publication is not guaranteed.'
 ]}
(OUT/'reviewer_response_status.json').write_text(json.dumps(readiness,indent=2))
print('Saved reviewer-revised outputs:',OUT)
display(pair_pre);display(pair_post.head(20))
```
